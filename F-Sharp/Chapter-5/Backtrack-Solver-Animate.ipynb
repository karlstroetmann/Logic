# A Backtracking Solver for CSPs with Animation

This notebook contains a variant of the backtracking solver from the notebook `02-Backtracking-Constraint-Solver.ipynb`.  This variant displays every partial assignment that it tries.  Furthermore, it counts the number of partial assignments that have been tested.

We import the types `Constraint<'V>` and `CSP<'V>`, the functions to create constraints, and the function `isConsistent` from the notebook `02-Backtracking-Constraint-Solver.ipynb`.

```
importNotebook "02-Backtracking-Constraint-Solver.ipynb"
```

## The Backtracking Solver with Animation

The mutable variable `count` counts the number of partial assignments that have been tested.

```
let mutable count = 0
```

The function `backtrackSearchAnimated` takes three arguments:
- `show` is a function that displays a partial variable assignment, 
- `Assignment` is a partial variable assignment that is represented as a map.  Initially, this assignment will be the empty map.  Every recursive call of `backtrackSearchAnimated` adds the assignment of one variable to the given assignment.  The important invariant of recursive calls is that `Assignment` is *consistent*, i.e. all constraints $f$ that contain only variables from the set $\mathtt{dom}(\mathtt{Assignment})$ are satisfied.
- `csp` is a constraint satisfaction problem.

The function tries to find a solution of `csp` by recursively augmenting `Assignment`.  It works like the function `backtrackSearch` from the notebook `02-Backtracking-Constraint-Solver.ipynb`, but every partial assignment is displayed using the function `show`, and the number of tested assignments is counted.

```
let rec backtrackSearchAnimated (show: Map<string, 'V> -> unit) (Assignment: Map<string, 'V>) (csp: CSP<'V>) : Map<string, 'V> option =
    show Assignment
    if Assignment.Count = csp.Variables.Length then
        Some Assignment
    else
        let var = csp.Variables |> List.find (fun x -> not (Assignment.ContainsKey x))
        csp.Values |> List.tryPick (fun value ->
            count <- count + 1
            if isConsistent var value Assignment csp.Constraints then
                backtrackSearchAnimated show (Assignment.Add(var, value)) csp
            else
                None)
```

The function `solveAnimated` takes two arguments:
- `show` is a function that displays a partial variable assignment,
- `csp` is a constraint satisfaction problem.

It solves `csp` via backtracking.  Every partial assignment that is tried is displayed using `show`.  Finally, the number of tested assignments is printed and the solution is returned.

```
let solveAnimated (show: Map<string, 'V> -> unit) (csp: CSP<'V>) : Map<string, 'V> option =
    count <- 0
    let Solution = backtrackSearchAnimated show Map.empty csp
    printfn "tested %d partial assignments" count
    Solution
```
