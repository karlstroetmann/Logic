# Odd Perfect Numbers

A natural number $n$ is perfect if $n$ is equal to the sum of its proper divisors:
$$ \sum \bigl\{t \mid t \in \{1,\cdots, n-1\} \wedge n \;\%\; t = 0 \bigr\}  = n $$
For example, $6$ is perfect, because $6 = 1 + 2 + 3$, but $8$ is not perfect, because 
$1 + 2 + 4 = 7 \not= 8$.

The function `perfect` takes one argument:
- `n` is a positive natural number.

It returns `true` if `n` is perfect.  `List.sum` computes the sum of the elements of a list.

```
let perfect (n: int) : bool =
    List.sum [ for t in 1 .. n - 1 do if n % t = 0 then t ] = n
```

```
perfect 6
```

```
perfect 8
```

The function `findOddPerfect` takes no arguments.  It searches for an odd perfect number by testing the odd numbers $1, 3, 5, \cdots$ one after the other.  If an odd perfect number is found, it is returned.  Every 1000 numbers, the current number is printed, so that we can watch the progress of the search.

It is unknown whether there are odd perfect numbers.  Hence it is unknown whether this function terminates.

```
let findOddPerfect () : int =
    let rec search (n: int) =
        if perfect n then
            n
        else
            if n % 1000 = 1 then
                printfn "%d" n
            search (n + 2)
    search 1
```
