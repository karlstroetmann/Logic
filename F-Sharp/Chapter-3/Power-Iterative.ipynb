# Iterative Computation of Powers via Iterative Squaring

The function `power` takes two arguments:
- `x` is a natural number, the *base*,
- `y` is a natural number, the *exponent*.

The function returns the power $x^y$.  As powers get very big, the base `x` and the result have the type `bigint`, i.e. they are integers of arbitrary size.  Literals of type `bigint` are written with the suffix `I`, e.g. `1I`.

**Implementation:** As the parameters of a function can not be changed in F#, we use the *mutable* variables `x`, `y`, and `r`.  The variables `x` and `y` are initialized with the values of the parameters of the same name.  The function maintains the *invariant*
$$ r_n \cdot x_n^{y_n} = x_0^{y_0}. $$
Here $x_n$, $y_n$, and $r_n$ are the values of the variables `x`, `y`, and `r` after $n$ iterations of the loop.  Note that we have:
* $x_{n+1} = x_n^2$,
* $y_{n+1} = y_n \;\texttt{/}\; 2$, 
* $r_{n+1} = r_n \cdot x_n$ if $y_n \;\texttt{%}\; 2 = 1$, and
* $r_{n+1} = r_n$ if $y_n \;\texttt{%}\; 2 = 0$.

Here, $y_n \;\texttt{/}\; 2$ denotes the integer division of $y_n$ by $2$.
When the loop terminates, we have $y_n = 0$ and hence the invariant implies $r_n = x_0^{y_0}$.

```
let power (x: bigint) (y: int) : bigint =
    let mutable x = x
    let mutable y = y
    let mutable r = 1I
    while y > 0 do
        if y % 2 = 1 then
            r <- r * x
        x <- x * x
        y <- y / 2
    r
```

We compute $3^{500\,000}$.  This number has more than $200\,000$ digits.  Therefore, we only display the number of its digits.

```
#!time
let p = power 3I 500_000
```

We test the function `power` using random numbers.  We compare it with the function `pown` from the F# library.
`System.Random()` creates a random number generator and `rnd.Next(a, b)` returns a random number $x$ such that $a \leq x < b$.

```
let rnd = System.Random()
```

```
#!time
for _ in 1 .. 100_000 do
    let a = rnd.Next(1, 1001)
    let b = rnd.Next(0, 1001)
    if power (bigint a) b <> pown (bigint a) b then
        failwithf "Error: power %d %d" a b
printfn "All tests have been successful!"
```
