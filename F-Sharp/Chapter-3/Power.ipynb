# Computation of Powers via Iterative Squaring

The function `power` takes two arguments:
- `m` is a natural number, the *base*,
- `n` is a natural number, the *exponent*.

The function returns the power $m^n$.  As powers get very big, the base `m` and the result have the type `bigint`, i.e. they are integers of arbitrary size.  Literals of type `bigint` are written with the suffix `I`, e.g. `1I`.

**Implementation:** The *recursive implementation* of the function `power` given below is based on the following two equations:
1. $m^0 = 1$
2. $m^n = \left\{\begin{array}{ll}
         m^{n/2} \cdot m^{n/2}          & \mbox{if $n$ is even};    \\
         m^{n/2} \cdot m^{n/2} \cdot m  & \mbox{if $n$ is odd}.
         \end{array}
         \right.
         $

Here, $n/2$ denotes the integer division of $n$ by $2$, which is computed by the operator `/` in F$^\sharp$.
Effectively, we square the $m^{n/2}$.  Hence this method is also known as *iterative squaring*.

```
let rec power (m: bigint) (n: int) : bigint =
    if n = 0 then
        1I
    else
        let p = power m (n / 2)
        if n % 2 = 0 then
            p * p
        else
            p * p * m
```

We compute $3^{500\,000}$.  This number has more than $200\,000$ digits.  Therefore, we only display the number of its digits.

```
#!time
let p = power 3I 500_000
```

```
(string p).Length
```

We compare the result with the result of the function `pown` from the F$^\sharp$ library.

```
pown 3I 500_000 = power 3I 500_000
```
