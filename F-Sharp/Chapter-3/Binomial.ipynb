# Recursive Computation of the Binomial Coefficients

The *binomial coefficient* ${n \choose k}$ is defined as follows:
$$ {n \choose k} = \frac{n!}{k! \cdot (n-k)!} $$
The function `binomial` takes two arguments:
- `n` is a natural number,
- `k` is a natural number such that $k \leq n$.

The function returns the binomial coefficient ${n \choose k}$.

**Implementation:** The function is based on the following recursive equations:
1. ${n \choose 0} = 1$ and ${n \choose n} = 1$,
2. ${n \choose k} = {n-1 \choose k-1} + {n-1 \choose k}$ if $0 < k < n$.

```
let rec binomial (n: int) (k: int) : int =
    if k = 0 || k = n then
        1
    else
        binomial (n - 1) (k - 1) + binomial (n - 1) k
```

In order to test the function `binomial`, we compute the binomial coefficients directly from their definition.
The function `factorial` takes one argument:
- `n` is a natural number.

It returns $n!$, i.e. the product $1 \cdot 2 \cdot \cdots \cdot n$.  As factorials grow very fast, the result is computed as a `bigint`, i.e. as an integer of arbitrary size.  The literal `1I` denotes the number $1$ as a `bigint`, the function `List.fold` multiplies all numbers in the list `[1I .. bigint n]`, starting with the value `1I`.

```
let factorial (n: int) : bigint =
    List.fold (*) 1I [1I .. bigint n]
```

The function `comb` takes two arguments:
- `n` is a natural number,
- `k` is a natural number such that $k \leq n$.

It returns ${n \choose k}$ computed via the formula $\frac{n!}{k! \cdot (n-k)!}$.  The result is converted from `bigint` to `int`.

```
let comb (n: int) (k: int) : int =
    int (factorial n / (factorial k * factorial (n - k)))
```

We check that both functions compute the same values.  F# has a function `assert`, but it is only active when the code is compiled in *debug mode*.  Therefore, we raise an exception explicitly if a test fails.  The function `failwithf` raises an exception with an error message that is formatted like the output of `printfn`.

```
for n in 0 .. 11 do
    for k in 0 .. n do
        if binomial n k <> comb n k then
            failwithf "Error: binomial %d %d = %d" n k (binomial n k)
printfn "All tests have been successful!"
```

Finally, we print *Pascal's triangle*.

```
for n in 1 .. 9 do
    for k in 0 .. n do
        printf "%d " (binomial n k)
    printfn ""
```
