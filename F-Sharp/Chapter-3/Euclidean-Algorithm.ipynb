# The Euclidean Algorithm for Computing the Greatest Common Divisor 

The function `gcd` takes two arguments:
- `a` is a natural number,
- `b` is a natural number.

The function returns the *greatest common divisor* of `a` and `b`.

**Implementation:** The function is based on the equations $\texttt{gcd}(a, 0) = a$ and $\texttt{gcd}(a, b) = \texttt{gcd}(b, a \,\texttt{%}\, b)$ for $b \not= 0$.  As long as $b \not= 0$, the pair $(a, b)$ is replaced by the pair $(b, a \,\texttt{%}\, b)$.  This does not change the greatest common divisor.  

In F#, the parameters of a function can not be changed.  Therefore, we introduce the *mutable* variables `x` and `y` that are initialized with `a` and `b`.  A mutable variable is changed using the operator `<-`.  Since `F#` does not have simultaneous assignment like `a, b = b, a % b` in Python, we need the auxiliary variable `r` to store the remainder.

```
let gcd (a: int) (b: int) : int =
    let mutable x = a
    let mutable y = b
    while y <> 0 do
        let r = x % y
        x <- y
        y <- r
    x
```

```
gcd 125 20
```
