# Algebrite Function Overview
This notebook provides a quick overview of the most common functions available in the Algebrite computer algebra system. Each core feature is demonstrated using its TypeScript definitions.

```typescript
import * as Algebrite from 'algebrite';
```

##  `run(expression)`
The `run` function is the simplest way to interact with Algebrite. It takes a string containing an algebraic expression or command, evaluates it, and directly returns the result as a string.

```typescript
const runResult = Algebrite.run('x + x');
console.log('run(x + x) =', runResult); // Expected: 2 x
```

##  `simplify(expression)`
Simplifies the given mathematical expression. For programmatic methods like this (unlike `run`), you need to call `.toString()` to extract the readable string representation of the resulting object.

```typescript
const simplified = Algebrite.simplify('2*x + 3*x').toString();
console.log('simplify(2*x + 3*x) =', simplified); 
```

##  `factor(expression)`
Factors a polynomial expression into its constituent multiplicative terms.

```typescript
const factored = Algebrite.factor('x^2 - y^2').toString();
console.log('factor(x^2 - y^2) =', factored); 
```

##  `expand(expression)`
Expands a factored expression or power into a full polynomial.

```typescript
const expanded = Algebrite.expand('(x+y)^2').toString();
console.log('expand((x+y)^2) =', expanded); 
```

##  `subst(expression, oldVar, newVar)`
Substitutes a variable or sub-expression within an equation with a new value or expression.

```typescript
const substituted = Algebrite.subst('3', 'x', 'x^2').toString();
console.log('subst(3, x, x^2) =', substituted); 
```

##  `derivative(expression)`
Computes the symbolic derivative of the given expression with respect to the primary variable (usually `x`).

```typescript
const deriv = Algebrite.derivative('x * sin(x)').toString();
console.log('derivative(x * sin(x)) =', deriv); 
```

##  `integral(expression)`
Computes the symbolic indefinite integral (antiderivative) of the given expression.

```typescript
const integ = Algebrite.integral('x*sin(x)').toString();
console.log('integral(x*sin(x) =', integ); 
```

##  `roots(expression)`
Attempts to find the roots (the points where the expression evaluates to zero) of a given polynomial.

```typescript
const rootsResult = Algebrite.roots('x^2 - 2').toString();
console.log('roots(x^2 - 4) =', rootsResult); 
```
