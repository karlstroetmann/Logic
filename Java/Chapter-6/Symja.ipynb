# Symja Function Overview
This notebook provides a quick overview of the most common functions available in the computer algebra system [Symja](https://github.com/axkr/symja_android_library).  Symja is implemented in Java.  It is the Java counterpart of the JavaScript library *Algebrite*, which is used in the TypeScript version of these notebooks.

Symja uses a syntax that is similar to the syntax of *Mathematica*: The names of built-in functions start with an upper case letter and their arguments are enclosed in parentheses.

Symja is available from *Maven Central*.  The first time the following command is executed, it downloads Symja and the libraries it depends on.  This takes about a minute.

```java
%maven org.matheclipse:matheclipse-core:3.2.0
```

Symja uses logging libraries that print warnings if they are not configured.  The following two system properties suppress these warnings.  They have to be set before Symja is used for the first time.

```java
System.setProperty("slf4j.internal.verbosity", "ERROR");
System.setProperty("log4j2.statusLoggerLevel", "OFF");
```

```java
import org.matheclipse.core.eval.ExprEvaluator;
```

##  `eval(expression)`
An object of the class `ExprEvaluator` evaluates expressions.  Its method `eval` takes a string containing an algebraic expression or command, evaluates it, and returns the result.  The result is an object of type `IExpr`.  Its method `toString()` returns the result as a string.

```java
var util = new ExprEvaluator();
```

```java
var evalResult = util.eval("x + x");
System.out.println("eval(x + x) = " + evalResult);  // Expected: 2*x
```

##  `Simplify(expression)`
Simplifies the given mathematical expression.

```java
var simplified = util.eval("Simplify(2*x + 3*x)").toString();
System.out.println("Simplify(2*x + 3*x) = " + simplified);
```

##  `Factor(expression)`
Factors a polynomial expression into its constituent multiplicative terms.

```java
var factored = util.eval("Factor(x^2 - y^2)").toString();
System.out.println("Factor(x^2 - y^2) = " + factored);
```

##  `Expand(expression)`
Expands a factored expression or power into a full polynomial.

```java
var expanded = util.eval("Expand((x+y)^2)").toString();
System.out.println("Expand((x+y)^2) = " + expanded);
```

##  `ReplaceAll(expression, oldVar -> newVar)`
Substitutes a variable or sub-expression within an expression with a new value or expression.  In Algebrite, this function is called `subst`.

```java
var substituted = util.eval("ReplaceAll(x^2, x -> 3)").toString();
System.out.println("ReplaceAll(x^2, x -> 3) = " + substituted);
```

##  `D(expression, variable)`
Computes the symbolic derivative of the given expression with respect to the given variable.

```java
var deriv = util.eval("D(x * Sin(x), x)").toString();
System.out.println("D(x * Sin(x), x) = " + deriv);
```

##  `Integrate(expression, variable)`
Computes the symbolic indefinite integral (antiderivative) of the given expression.

```java
var integ = util.eval("Integrate(x*Sin(x), x)").toString();
System.out.println("Integrate(x*Sin(x), x) = " + integ);
```

##  `Roots(equation, variable)` and `Solve(equation, variable)`
The function `Roots` finds the roots of a polynomial equation, i.e. the points where the polynomial evaluates to zero.  The function `Solve` returns the solutions as a list of substitutions.

```java
var rootsResult = util.eval("Roots(x^2 - 2 == 0, x)").toString();
System.out.println("Roots(x^2 - 2 == 0, x) = " + rootsResult);
```

```java
var solveResult = util.eval("Solve(x^2 - 2 == 0, x)").toString();
System.out.println("Solve(x^2 - 2 == 0, x) = " + solveResult);
```
