## Basic Concepts: Types and Methods

Before introducing the individual data types and language constructs of Java, this section introduces two basic concepts that will recur throughout the chapter: types and methods.  In addition, the keywords `var` and `final` for variable declarations are introduced.

**Types**

Java is a *statically typed* language: Every variable, every parameter, and the return value of every method has a type that is known at compile time.  The type is written *in front of* the name of the variable:

```java
String  name   = "Alice"; // Variable of type String
int     age    = 25;      // Variable of type int
boolean active = true;    // Variable of type boolean
```

Java also supports *type inference* for local variables: If a variable is declared with the keyword `var` and a value is assigned directly during the declaration, the compiler derives the type of the variable from this value.  For example, when we define the variable `email` below, Java infers that the type of `email` is `String`.

```java
var email = "karl.stroetmann@dhbw.de";
```

Therefore, we cannot assign a number to `email`.  The next cell yields an error.

```java
email = 42
```

**Variables that cannot be changed: `final`**

If a variable is declared with the keyword `final`, it cannot be changed after its initialization.  Variables are *block-scoped*, meaning they are only valid within the block (e.g., a method, a loop, or an `if` block) in which they were declared.  A block is enclosed in curly braces.

*JShell* ignores the keyword `final` for variables that are defined at the top level of a cell.  Therefore, we demonstrate `final` inside a block.  The next cell yields an error.

```java
{
    final double pi = 3.14159265;  // immutable
    pi = 3;                        // Error: cannot assign a value to final variable pi
}
```

```java
int counter = 0;  // mutable
counter = 1;      // allowed
```

The types of the parameters as well as the return type have to be specified.  The return type is written in front of the name of the method. **Lambda Expressions**

In addition to methods, Java supports *lambda expressions*, which are anonymous functions.  To the left of the arrow (`->`) are the parameters, and to the right is the return expression.  A lambda expression has no type of its own.  It can be assigned to a variable whose type is a *functional interface*, i.e. an interface with exactly one abstract method.  The package `java.util.function` contains many functional interfaces.  For example, `BinaryOperator<Integer>` is the type of functions that take two arguments of type `Integer` and return an `Integer`.  Its method `apply` calls the function.

```java
BinaryOperator<Integer> sub = (a, b) -> a - b;
```

```java
BinaryOperator<Integer> mul = (a, b) -> {
    int product = a * b;
    return product;
};
mul.apply(3, 5)
```

Lambda expressions are often used for short functions and are particularly used in combination with the stream methods `filter`, `map`, and `reduce`, which are introduced in the section on functional programming.

## Primitive Data Types and Arithmetic

In TypeScript and JavaScript, all numbers are floating-point numbers.  Unlike these languages, Java distinguishes between integers and floating-point numbers.  It has eight *primitive* data types:

| **Type** | **Values** |
| :--- | :--- |
| `byte`, `short`, `int`, `long` | integers with 8, 16, 32, and 64 bits |
| `float`, `double` | floating-point numbers with 32 and 64 bits according to the [IEEE-754 standard](https://en.wikipedia.org/wiki/IEEE_754) |
| `char` | characters, e.g. `'a'` |
| `boolean` | the truth values `true` and `false` |

Values of a primitive type are not objects.  For every primitive type there is a *wrapper class*, e.g. `Integer` for `int` and `Double` for `double`.  If a primitive value is converted to an `Object`, it is automatically wrapped into an object of the wrapper class.  The method `getClass()` returns the class of an object at runtime.

```java
((Object) 42).getClass()
```

```java
((Object) 42.123).getClass()
```

The range of every integer type is limited.  For example, the largest value of type `int` is $2^{31} - 1$.  If this limit is exceeded, the result *wraps around* silently: This is called an *overflow*.

```java
Integer.MAX_VALUE
```

```java
Integer.MAX_VALUE + 1
```

```java
Long.MAX_VALUE
```

The precision of the type `double` is limited, too.  Integers are only represented exactly as long as they are not bigger than $2^{53}$.  Beyond this limit, rounding errors occur, as shown below.

```java
9007199254740992.0 == 9007199254740993.0
```

As `BigInteger` is a class, there are no literals for this type.  Instead, the static method `BigInteger.valueOf` converts a `long` into a `BigInteger`.  As Java does not support *operator overloading*, the arithmetic operations are methods: `add`, `subtract`, `multiply`, `divide`, `mod`, and `pow`.

## Functional Programming with Streams

A *stream* is a sequence of elements that supports *functional programming*.  The method `stream()` turns a list into a stream, while the method `toList()` collects the elements of a stream in a list.  Streams provide the following methods:
* `filter`,
* `map`, 
* `reduce`, 
* `anyMatch`, and `allMatch`. 

**The `filter` Method**

The `filter` method creates a new stream that contains only those elements of the original stream for which a passed condition is met. The original list is not modified. The condition is passed as a function that returns either `true` (element is kept) or `false` (element is discarded) for each element. Usually, a *lambda expression* is used for such short functions.

In the following example, only numbers greater than 3 are filtered out of a list of numbers:

```java
var numbers = List.of(1, 2, 3, 4, 5);
var result  = numbers.stream().filter(n -> n > 3).toList();
result
```

The function `n -> n > 3` is called internally for each element of the list. For `n = 1`, `2`, and `3`, it returns `false`, which is why these elements are discarded. For `n = 4` and `5`, it returns `true`, so they remain in the result. The original list `numbers` remains unchanged.

| **Element (n)** | **Condition (n > 3)** | **Result** |
| :--- | :--- | :--- |
| 1 | false | Drop |
| 2 | false | Drop |
| 3 | false | Drop |
| 4 | true | Keep |
| 5 | true | Keep |

**The `map` Method**

The `map` method transforms each element of a stream using a passed function and returns a new stream containing the transformed values. The length of the resulting stream is always identical to that of the original stream.

In the following example, each element of a list of numbers is doubled:

```java
var numbers = List.of(1, 2, 3, 4);
var result  = numbers.stream().map(n -> n * 2).toList(); 
result
```

The function `n -> n * 2` is called once for each element and returns its doubled value. The result is a new list where each position contains the transformed value of the corresponding position in the original. The original list `numbers` remains unchanged.

**The `reduce` Method**

The `reduce` method condenses a stream into a single value by applying a function to each element along with a running intermediate result (the so-called *accumulator*).

The call `numbers.stream().reduce(0, (acc, n) -> acc + n)` consists of the following parts:
1. The **initial value**: The `0` is the initial value for the calculation.
2. The **combining function**: `(acc, n) -> acc + n` defines the calculation rule.
3. The **parameters**:
    * `acc` (short for *Accumulator*): Stores the intermediate result of all previous additions.
    * `n`: Is the current element from the stream that is currently being processed.

Note that the initial value is the *first* argument of `reduce`.

In the following example, the sum of all elements of a list is calculated:

```java
var numbers = List.of(1, 2, 3, 4);
var sum     = numbers.stream().reduce(0, (acc, n) -> acc + n);
sum
```

The accumulator `acc` starts with the value `0`. In each step, the current element `n` is added to the previous accumulator value, and the result is passed on as the new accumulator value:

| **Step** | **Accumulator (acc)** | **Element (n)** | **New acc** |
| :--- | :--- | :--- | :--- |
| Start | 0 | --- | 0 |
| 1 | 0 | 1 | 1 |
| 2 | 1 | 2 | 3 |
| 3 | 3 | 3 | 6 |
| 4 | 6 | 4 | 10 |

**The `anyMatch` Method**

The method `anyMatch` checks whether there is at least one element that satisfies a given condition. For example,
below we check whether `numbers` contains a number divisible by `13`.

```java
var numbers = List.of(15, 30, 60, 39);
numbers.stream().anyMatch(n -> n % 13 == 0)
```

**The `allMatch` Method**

The method `allMatch` checks whether all elements of a stream satisfy a given condition. For example,
below we check whether all numbers contained in `numbers` are even.

```java
var numbers = List.of(2, 4, 6, 8, 10);
numbers.stream().allMatch(n -> n % 2 == 0)
```

**Creating Lists with `IntStream`**

Besides listing the elements with `List.of`, lists can be created from streams.  This is particularly useful when a list of a specific length needs to be populated with calculated values, for example, to generate a sequence of numbers.

The class `IntStream` represents streams of values of type `int`.  The method `IntStream.range(a, b)` creates the stream of the numbers from `a` up to but excluding `b`, while `IntStream.rangeClosed(a, b)` includes `b`.  In the following example, a list with the numbers 1 to 10 is created:

```java
IntStream.rangeClosed(1, 10).boxed().toList()
```

The method calls are interpreted as follows:
1. `IntStream.rangeClosed(1, 10)` creates the stream of the numbers from 1 up to and including 10.
2. `boxed()` converts every value of type `int` into an object of type `Integer`, since a list can only store objects.
3. `toList()` collects the elements of the stream in a list.

The values can be transformed with `map`.  For example, the following expression computes the list of the first 10 square numbers:

```java
IntStream.range(0, 10).map(i -> (i + 1) * (i + 1)).boxed().toList()
```

**Head and Tail of a List**

Java does not support *array destructuring*.  In order to separate the *head* and the *tail* of a list, a common pattern in functional programming, we use the methods `getFirst` and `subList`.  The call `list.subList(a, b)` returns the list of the elements with an index from `a` up to but excluding `b`.

```java
var list = List.of(10, 20, 30, 40, 50);
var head = list.getFirst();
var tail = list.subList(1, list.size());
System.out.println(head); 
System.out.println(tail);
```

## Boolean Values and Operators

The representation of truth values in Java is done via the primitive data type `boolean`, which can only take the values `true` and `false`. These values are central for controlling the program flow and for logical decisions.

**Logical Operators and Lazy Evaluation**

Java supports the classic logical operations conjunction (AND), disjunction (OR), and negation (NOT):
* `a && b` is the conjunction of `a` and `b`,
* `a || b` is the disjunction of `a` and `b`, and
* `!a` is the negation of `a`.

An essential concept in the evaluation of `&&` and `||` is *lazy evaluation* (often referred to as *short-circuit evaluation* in this context). This means that the second operand is only evaluated if it is absolutely necessary for the final result:
* For `A && B`, `B` is no longer evaluated if `A` is already `false` (the result is definitely `false` anyway).
* For `A || B`, `B` is no longer evaluated if `A` is already `true` (the result is definitely `true` anyway).

The operators `&` and `|` also compute the conjunction and the disjunction, but they always evaluate both operands.  The operator `^` computes the *exclusive or*.

```java
true && false
```

```java
true || false
```

```java
!true
```

```java
true ^ true
```

**Comparison Operators and Equality**

Comparisons are frequently used to generate truth values.  The operators `==` and `!=` check whether two values are equal or different.  Java does not perform an automatic conversion between strings and numbers.  Hence, the expression `5 == "5"` is not *well-typed* and is rejected by the Java compiler.

```java
5 == "5"
```

For *objects*, the operator `==` checks whether both operands refer to the *same* object.  In order to check whether two objects have the same *content*, the method `equals` has to be used.  For example, the following expression creates a new string with the content `"abc"`.  This string is a different object than the string literal `"abc"`.

```java
new String("abc") == "abc"
```

```java
new String("abc").equals("abc")
```

This is a common source of errors, as objects of the wrapper class `Integer` are objects, too.  The comparison of two objects of type `Integer` with `==` might yield `false` although they represent the same number.

```java
Integer a = 1000, b = 1000;
a == b
```

```java
a.equals(b)
```

The classic size comparisons (`<`, `>`, `<=`, `>=`) are available in Java.

An important difference compared to Python is the chaining of comparisons. A mathematical expression like $3 > 2 > 1$ is not accepted in Java, because the result of `3 > 2` is a `boolean` and we can't compare a `boolean` with a number.  Therefore, mathematical range checks must be explicitly linked with the *logical and operator*: `3 > 2 && 2 > 1`.

```java
3 > 2 > 1
```

```java
3 > 2 && 2 > 1
```

## Control Structures

Control structures serve to control the flow of a program.  Java offers classic constructs for this purpose, which are syntactically based on the programming language C.

**Branches**

The basic conditional execution is done using `if`, `else if`, and `else`. Unlike Python, the conditions must be placed in parentheses `()` and they must have the type `boolean`. The associated instruction blocks are usually enclosed in curly braces `{}`. Although Java syntax allows omitting the curly braces if the block consists of exactly one single instruction, explicit bracing is considered a best practice for readability and error prevention (e.g., when adding further instructions later).

```java
int x = 17;
if (x < 0) {
    System.out.println("The number is negative!");
} else if (x == 0) {
    System.out.println("The number is zero.");
} else {
    System.out.println("The number is positive.");
}
```

Alternatively, the `switch` statement offers a structured way to compare a value with several constants. At the end of each `case` block, the `break` keyword is used to force an exit from the `switch` structure. If `break` is missing, the program "falls through" into the next `case`. 

```java
int x = 0;
```

```java
switch (x) {
    case 0:
        System.out.println("The number is zero.");
        break;
    case 1:
        System.out.println("The number is one.");
        break;
    default:
        System.out.println("The number is something else.");
}
```

Modern Java also supports `switch` *expressions*, which use arrows `->` instead of colons.  These branches do not fall through, so no `break` is needed.  A `switch` expression returns a value.  The branches can contain *patterns* with conditions, which are introduced with the keyword `when`.  The pattern `Integer n when n < 0` matches every integer `n` that is negative.

```java
Integer y = -5;
String description = switch (y) {
    case 0                    -> "The number is zero.";
    case Integer n when n < 0 -> "The number is negative!";
    default                   -> "The number is positive.";
};
description
```

**Loop Constructs**

Java provides various loop constructs for iteration:
* The classic **`for` loop** uses a counter variable, a running condition, and a step size. It offers maximum control over the iteration process.

```java
for (int i = 0; i < 3; i++) {
    System.out.println(i);
}
```

* The **enhanced `for` loop** is ideally suited for iterating directly over the elements of an array, a list, or a set. It is more compact and less error-prone than the classic index-based `for` loop.

```java
var list = List.of(1, 2, 3);
for (int element : list) {
    System.out.println(element);
}
```

* The **`while` loop** executes a block of code as long as a defined running condition is met.

```java
var list = List.of(1, 2, 3);
int i = 0;
while (i < list.size()) {
    System.out.println(list.get(i));
    i++;
}
```

Inside loops, the `continue` keyword can be used to prematurely terminate the current loop iteration and continue with the next iteration.

```java
int[] arr = {1, 2, 3};
for (int i = 0; i < arr.length; ++i) {
    if (i == 1) {
        continue;
    }
    System.out.println("arr[" + i + "] = " + arr[i]);
}
```

**Application Example: [Sieve of Eratosthenes](https://en.wikipedia.org/wiki/Sieve_of_Eratosthenes)**

A classic algorithm for determining all prime numbers up to a given limit $n\in\mathbb{N}$ is the Sieve of Eratosthenes. The implementation in Java demonstrates the interaction of arrays, the classic `for` loop as well as the `while` loop, and conditional statements:

1. Initialization of an array `primes`, where `primes[i] = i`.
2. Iteration over all numbers $i$ and $j$ to eliminate multiples (i.e., set them to 0).
3. Filtering the remaining numbers $\neq 0$.

An optimized version also uses `Math.ceil(Math.sqrt(n))` ($\lceil \sqrt{n} \rceil$) as the upper limit for the outer loop and starts the inner loop at $j=i$ to avoid redundant checks.  As `Math.ceil` returns a `double`, the result is converted into an `int` using the type cast `(int)`.  The method `Arrays.stream` turns an array of type `int[]` into an `IntStream`.

```java
int[] primes(int n) {
    // Initialize array with values 0 to n
    int[] primes = IntStream.rangeClosed(0, n).toArray();
    int   limit  = (int) Math.ceil(Math.sqrt(n));
    for (int i = 2; i <= limit; i++) {
        // If i is already marked (0), skip
        if (primes[i] == 0) {
            continue; 
        }
        int j = i;
        // Cross out all multiples of i (starting at i*i)
        while (i * j <= n) {
            primes[i * j] = 0;
            ++j;
        }
    }
    // Filter all non-zero values (prime numbers)
    return Arrays.stream(primes).filter(p -> p >= 2).toArray();
}
```

Below, we need to use `Arrays.toString` to see the array.

```java
Arrays.toString(primes(10_000))
```

**Rounding Functions**

Three specific methods are available for rounding floating-point numbers to integers:
1. **`Math.floor(x)`:** The *floor function* $\lfloor x \rfloor$. It always rounds down to the next smaller integer.  Note that the result is a `double`.
$$ \text{floor}(x) = \max \{ n \in \mathbb{Z} \mid n \le x \} $$

```java
Math.floor(4.8)
```

2. **`Math.ceil(x)`:** The *ceiling function* $\lceil x \rceil$. It always rounds up to the next larger integer.  The result is a `double`, too.
$$ \text{ceil}(x) = \min \{ n \in \mathbb{Z} \mid n \ge x \} $$

```java
Math.ceil(4.2)
```

3. **`Math.round(x)`:** Commercial rounding to the nearest integer. Here, it rounds up starting from $.5$.  The result is a `long`.

```java
Math.round(4.5)
```

```java
Math.round(4.49)
```
