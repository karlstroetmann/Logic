# The Euclidean Algorithm for Computing the Greatest Common Divisor 

The function `ggt` takes two natural numbers `a` and `b` as arguments and computes their *greatest common divisor*.  As Java has no simultaneous assignment, we need the auxiliary variable `r` to replace `a` by `b` and `b` by `a % b`.

```java
long ggt(long a, long b) {
    while (b != 0) {
        long r = a % b;
        a = b;
        b = r;
    }
    return a;
}
```

```java
ggt(120, 20)
```
