# Euclidean Division

The function `divMod` takes two natural numbers $m$ and $n$ as arguments.
We must have $n > 0$.

It returns the unique pair $(q, r)$
that satisfies the following formulas:
 * $m = q \cdot n + r$,
 * $r < n$.

The pair is represented as an object of the record class `QR`.  Note that in Java the operator `/` performs *integer division* if both operands are integers.

```java
record QR(long q, long r) {}
```

```java
QR divMod(long m, long n) {
    if (m < n) {
        return new QR(0, m);
    }
    var qr = divMod(m / 2, n);
    long q = qr.q();
    long r = qr.r();
    if (2 * r + (m % 2) < n) {
        return new QR(2 * q, 2 * r + (m % 2));
    } else {
        return new QR(2 * q + 1, 2 * r + (m % 2) - n);
    }
}
```

```java
divMod(23, 5)
```

We use the function `testDivMod` defined below to test `divMod` using random numbers.  The method `nextLong(min, max)` of the class `ThreadLocalRandom` returns a random number from the interval $[\texttt{min}, \texttt{max})$.

```java
void testDivMod() {
    var random = ThreadLocalRandom.current();
    for (int i = 0; i < 1_000_000; i++) {
        long m  = random.nextLong(0, 1_000_000_000_000L);
        long n  = random.nextLong(1,     1_000_000_000L);
        var  qr = divMod(m, n);
        long q  = qr.q();
        long r  = qr.r();
        if (m != q * n + r || r >= n) {
            throw new AssertionError("Failed for m=" + m + ", n=" + n + ", q=" + q + ", r=" + r);
        }
    }
    System.out.println("All tests have been successful!");
}
```

```java
testDivMod();
```
