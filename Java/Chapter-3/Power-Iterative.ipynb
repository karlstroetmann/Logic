# Iterative Computation of Powers via Iterative Squaring

The function `power(x, y)` computes $x^y$ for natural numbers `x` and `y`.
It maintains the *invariant*
$$ r_n \cdot x_n^{y_n} = x_0^{y_0}. $$
Here $x_n$, $y_n$, and $z_n$ are the values of the variables `x`, `y`, and `z` after $n$ iterations of the loop.  Note that we have:
* $x_{n+1} = x_n^2$,
* $y_{n+1} = \lfloor y_n \;\texttt{/}\; 2 \rfloor$, 
* $r_{n+1} = \left\{\begin{array}{ll}
                     r_n \cdot x_n & \texttt{if} \; y_n \% 2 = 1, \\
                     r_n           & \texttt{if} \; y_n \% 2 = 0. \\
                     \end{array}\right.
  $

As the powers of numbers get big fast, we use the class `BigInteger` for the base.

```java
BigInteger power(BigInteger x, int y) {
    BigInteger r = BigInteger.ONE;
    while (y > 0) {
        if (y % 2 == 1) {
            r = r.multiply(x);
        }
        x = x.multiply(x);
        y = y / 2;
    }
    return r;
}
```

```java
long start = System.currentTimeMillis();
var p = power(BigInteger.valueOf(3), 500_000);
System.out.println("Power(3^500_000): " + (System.currentTimeMillis() - start) + " ms");
```

We compare our implementation with the method `pow` of the class `BigInteger`.

```java
long start = System.currentTimeMillis();
for (int i = 0; i < 100_000; i++) {
    var a   = BigInteger.valueOf(random.nextInt(1_000) + 1);
    int b   = random.nextInt(1_000);
    var got = power(a, b);
    var exp = a.pow(b);
    if (!got.equals(exp)) {
        throw new AssertionError("Mismatch for a=" + a + ", b=" + b);
    }
}
System.out.println("random-checks: " + (System.currentTimeMillis() - start) + " ms");
System.out.println("All tests have been successful!");
```
