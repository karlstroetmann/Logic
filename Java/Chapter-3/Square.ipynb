# Recursive Computation of Squares

The function `square(n)` computes $n^2$ recursively using the equation $n^2 = (n-1)^2 + 2 \cdot n - 1$.

```java
int square(int n) {
    if (n == 0) {
        return 0;
    }
    return square(n - 1) + 2 * n - 1;
}
```

```java
for (int i = 0; i <= 11; ++i) {
    System.out.println("square(" + i + ") = " + square(i));
}
```
