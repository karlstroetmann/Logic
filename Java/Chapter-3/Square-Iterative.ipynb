# Iterative Computation of Squares

The function `square(s)` computes $s^2$ using only additions: It adds the first $s$ odd numbers.

```java
int square(int s) {
    int r = 0;
    int u = 1;
    while (s > 0) {
        r += u;
        u += 2;
        s -= 1;
    }
    return r;
}
```

```java
for (int i = 0; i <= 11; ++i) {
    System.out.println("square(" + i + ") = " + square(i));
}
```
