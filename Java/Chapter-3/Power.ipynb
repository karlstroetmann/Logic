# Computation of Powers via Iterative Squaring

The *recursive implementation* of the function `power` given below is based on the following two equations:
1. $m^0 = 1$
2. $m^n = \left\{\begin{array}{ll}
         m^{n/2} \cdot m^{n/2}                  & \mbox{if $n$ is even};    \\
         m^{(n-1)/2} \cdot m^{(n-1)/2} \cdot m  & \mbox{if $n$ is odd}.
         \end{array}
         \right.
         $
         
Observe that we square the result `p` of the recursive invocation of `power`. 
Hence this method is also known as *iterative squaring*.

As the powers of numbers get big fast, we use the class `BigInteger` for the base.
It would not make much sense to use this type for the exponent, since if we really have an exponent $n$ such that $n > 2^{31} - 1$, then $m^n$ would not fit into the computer memory unless $m$ is an element for the set $\{-1,0, 1\}$.  Therefore, the exponent has the type `int`. 

```java
BigInteger power(BigInteger m, int n) {
    if (n == 0) {
        return BigInteger.ONE;
    }
    BigInteger p = power(m, n / 2);
    if (n % 2 == 0) {
        return p.multiply(p);
    } else {
        return p.multiply(p).multiply(m);
    }
}
```

```java
for (int k = 0; k <= 10; ++k) {
    System.out.println(String.format("3^%2d = %6s", k, power(BigInteger.valueOf(3), k)));
}
```

```java
long start = System.currentTimeMillis();
var p = power(BigInteger.valueOf(3), 500_000);
System.out.println("power(3, 500_000): " + (System.currentTimeMillis() - start) + " ms");
```

```java
if (!BigInteger.valueOf(3).pow(500_000).equals(power(BigInteger.valueOf(3), 500_000))) {
    throw new AssertionError("power(3, 500_000) is wrong");
}
System.out.println("The test has been successful!");
```
