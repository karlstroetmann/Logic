# The Euclidean Algorithm for Computing the Greatest Common Divisor 

The function `ggt` takes two natural numbers `a` and `b` as arguments and computes their *greatest common divisor*.

```java
long ggt(long a, long b) {
    if (b == 0) {
        return a;
    }
    return ggt(b, a % b);
}
```

```java
ggt(120, 20)
```

```java
ggt(123456789, 24681012)
```
