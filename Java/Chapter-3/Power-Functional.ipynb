# A Functional Implementation of  `power`

```java
import java.util.stream.*;
```

The function `range` returns the list `[1, ..., n]`. 

```java
List<Integer> range(int n) {
    return IntStream.rangeClosed(1, n).boxed().toList();
}
```

The method `reduce` of a stream takes three arguments: the initial value, a function that combines the accumulated value with the next element of the stream, and a function that combines two accumulated values.  The last function is only needed when the stream is processed in parallel.

```java
BigInteger power(BigInteger m, int n) {
    return range(n).stream().reduce(BigInteger.ONE, (acc, current) -> acc.multiply(m), BigInteger::multiply);
}
```

```java
System.out.println(power(BigInteger.valueOf(3), 4));  // Output: 81
```
