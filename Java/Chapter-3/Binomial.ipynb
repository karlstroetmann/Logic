# Recursive Computation of the Binomial Coefficients

The *binomial coefficient* ${n \choose k}$ is defined as follows:
$$ {n \choose k} = \frac{n!}{k! \cdot (n-k)!} $$
The function `binomial(n, k)` returns ${n \choose k}$.

```java
int binomial(int n, int k) {
    if (k == 0 || k == n) {
        return 1;
    } else {
        return binomial(n - 1, k - 1) + binomial(n - 1, k);
    }
}
```

In order to test the function `binomial`, we compare it with the formula given above.  The function `factorial(n)` computes $n!$.  For $n < 12$, the result fits into a `long`.

```java
long factorial(int n) {
    long result = 1;
    for (int i = 2; i <= n; i++) {
        result *= i;
    }
    return result;
}
```

In Java, `assert` statements are only checked if the *Java virtual machine* is started with the option `-ea`.  Therefore, we throw an `AssertionError` explicitly if a test fails.

```java
for (int n = 0; n < 12; n++) {
    for (int k = 0; k <= n; k++) {
        if (binomial(n, k) != factorial(n) / (factorial(k) * factorial(n - k))) {
            throw new AssertionError("binomial(" + n + ", " + k + ") is wrong");
        }
    }
}
System.out.println("All tests have been successful!");
```

```java
for (int n = 1; n < 10; n++) {
    var line = new StringBuilder();
    for (int k = 0; k <= n; k++) {
        line.append(binomial(n, k)).append(" ");
    }
    System.out.println(line);
}
```
