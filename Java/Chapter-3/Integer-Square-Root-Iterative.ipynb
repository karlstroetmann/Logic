# Integer Square Root

The function `root(a)` takes a natural number $a$ that is less than $2^{128\,000}$ and returns the largest natural number $r$ such that
$r^2 \leq a$, i.e. we have
$$ \texttt{root}(a) := \max\bigl(\{ r \in \mathbb{N} \mid r^2 \leq a \}\bigr). $$
The implementation tries to compute the result bit by bit starting from the bit with the highest value.
Since $a < 2^{128\,000}$ we know that `root(a)` is less than $2^{64\,000}$.

As $a$ can be very large, we use the class `BigInteger` from the package `java.math`, which supports integers of arbitrary size.  As Java does not support operator overloading, the arithmetical operations are methods: `x.add(y)` computes $x + y$, `x.pow(2)` computes $x^2$, `x.divide(y)` computes the integer quotient of $x$ and $y$, and `x.compareTo(y)` returns a negative number, zero, or a positive number if $x < y$, $x = y$, or $x > y$, respectively.

```java
BigInteger root(BigInteger a) {
    if (a.compareTo(BigInteger.TWO.pow(128_000)) > 0) {
        throw new IllegalArgumentException("a must be less than 2^128_000");
    }
    BigInteger r = BigInteger.ZERO;
    BigInteger p = BigInteger.TWO.pow(64_000);
    while (p.signum() > 0) {
        if (a.compareTo(r.add(p).pow(2)) >= 0) {
            r = r.add(p);
        }
        p = p.divide(BigInteger.TWO);
    }
    return r;
}
```

```java
int numDigits = 10_000;
long start = System.currentTimeMillis();
var val = root(BigInteger.TWO.multiply(BigInteger.TEN.pow(2 * numDigits)));
var s   = val.toString();
System.out.println(s.charAt(0) + "." + s.substring(1));
System.out.println("Computation of " + numDigits + " digits of √2: " + (System.currentTimeMillis() - start) + " ms");
```

We can check this at [CATONMAT](https://catonmat.net/tools/generate-sqrt2-digits).

The class `BigInteger` provides the method `sqrt`, which computes the integer square root, too.  We use it to check our result.

```java
val.equals(BigInteger.TWO.multiply(BigInteger.TEN.pow(2 * numDigits)).sqrt())
```

When we compare this with the floating point implementation of the function `sqrt`, we can see that the floating point implementation has been rounded up in the last digit.

```java
Math.sqrt(2.0)
```
