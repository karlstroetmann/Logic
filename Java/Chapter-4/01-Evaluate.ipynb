## Representing Formulas in Java

In Java, a formula is an object of a class implementing the interface `Formula`.  For every propositional operator there is a *record class*:

| Formula            | Java Representation                 |
|:-------------------|:------------------------------------|
| $p$                | `new Var("p")`                      |
| $\top$             | `new Verum()`                       |
| $\bot$             | `new Falsum()`                      |
| $\neg f$           | `new Not(f)`                        |
| $f \wedge g$       | `new And(f, g)`                     |
| $f \vee g$         | `new Or(f, g)`                      |
| $f \rightarrow g$  | `new Implies(f, g)`                 |
| $f \leftrightarrow g$ | `new Equiv(f, g)`                |

Records automatically implement the methods `equals` and `hashCode` structurally.  We override the method `toString` so that formulas are printed in a readable form.  As the interface `Formula` does not define any methods, its definition is very short.

````java
interface Formula {}

record Var(String name) implements Formula {
    public String toString() { return name; }
}
record Verum() implements Formula {
    public String toString() { return "⊤"; }
}
record Falsum() implements Formula {
    public String toString() { return "⊥"; }
}
record Not(Formula arg) implements Formula {
    public String toString() { return "¬" + arg; }
}
record And(Formula lhs, Formula rhs) implements Formula {
    public String toString() { return "(" + lhs + " ∧ " + rhs + ")"; }
}
record Or(Formula lhs, Formula rhs) implements Formula {
    public String toString() { return "(" + lhs + " ∨ " + rhs + ")"; }
}
record Implies(Formula lhs, Formula rhs) implements Formula {
    public String toString() { return "(" + lhs + " → " + rhs + ")"; }
}
record Equiv(Formula lhs, Formula rhs) implements Formula {
    public String toString() { return "(" + lhs + " ↔ " + rhs + ")"; }
}
````

The method `toString` prints formulas *fully parenthesized*, so the structure of the tree is visible.

````java
Formula f0 = new Implies(new And(new Implies(new Var("p"), new Var("q")),
                                 new Implies(new Not(new Var("p")), new Var("q"))),
                         new Var("q"));
f0
````

    According to the conventions of our lecture, this string should be rejected as syntactically invalid.

Next, we need to decide how to represent a *propositional valuation*.  Mathematically, a *propositional valuation* $\mathcal{I}$ is a function of the form
$$ \mathcal{I}:\mathcal{P} \rightarrow \mathbb{B}  $$
that maps propositional variables to the set of Boolean values 
$\mathbb{B} = \{\texttt{true}, \texttt{false}\}$.  We can represent a propositional  valuation $\mathcal{I}$ as the set of those variables that are evaluated as <tt>true</tt> by $\mathcal{I}$:  
$$ \texttt{repr}(\mathcal{I}) := \bigl\{ v \in \mathcal{P} \mid \mathcal{I}(v) = \texttt{true} \bigr\}$$
For example, the following set would be a propositional valuation for the formula $\texttt{f}$ that interprets both `p` and `q` as `true`.

We use the class `RecursiveSet` to represent sets of variables.  We also define an abbreviation for its constructor.  The annotation `@SafeVarargs` suppresses a warning that is caused by the combination of *generics* and a variable number of arguments.

````java
@SafeVarargs
<T> RecursiveSet<T> set(T... elements) {
    return new RecursiveSet<>(elements);
}
````

````java
var I1 = set("p", "q");
I1
````

There are 4 possible propositional valuation for the formula $\texttt{f}$.  The other three propositional valuations are as follows:

````java
RecursiveSet<String> I2 = set("p");
RecursiveSet<String> I3 = set("q");
RecursiveSet<String> I4 = set();
````

The function `evaluate` defined below takes two arguments:
   - `f` is a formula, i.e. a tree of objects implementing the interface `Formula`.
   - `I` is a set of *propositional variables* representing a *propositional valuation*.

The function returns the value of the propositional formula `f` given the propositional valuation `I`.
It uses a `switch` expression with *record patterns*: The case
```java
    case And(var g, var h) -> ...
```
is chosen if `f` is a conjunction.  In this case, the variables `g` and `h` are bound to the left and right operand of this conjunction.

When evaluating a formula of the form $g \rightarrow h$ this function uses the fact that
$$ g \rightarrow h \Leftrightarrow \neg g \vee h. $$
In TypeScript we could have used the fact that `false < true`.  However, in Java the operator `<` can not be applied to Boolean values.

As the interface `Formula` might be implemented by other classes, too, the Java compiler requires a `default` case.  This case throws an exception.

````java
boolean evaluate(Formula f, RecursiveSet<String> I) {
    return switch (f) {
        case Var(var p)            -> I.contains(p);
        case Verum()               -> true;
        case Falsum()              -> false;
        case Not(var g)            -> !evaluate(g, I);
        case And(var g, var h)     -> evaluate(g, I) && evaluate(h, I);
        case Or(var g, var h)      -> evaluate(g, I) || evaluate(h, I);
        case Implies(var g, var h) -> !evaluate(g, I) || evaluate(h, I);
        case Equiv(var g, var h)   -> evaluate(g, I) == evaluate(h, I);
        default -> throw new IllegalArgumentException("unknown formula: " + f);
    };
}
````
