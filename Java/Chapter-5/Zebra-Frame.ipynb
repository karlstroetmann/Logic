The function $\texttt{zebraCSP}()$ returns a CSP that codes the zebra problem.  When implementing this function it is important to order the variables in a way that variables that are connected to each other by a constraint are tried in succession, for otherwise the CSP solver will take mu$\cdots$uch longer.

```java
var Nations = List.of("English", "Spaniard", "Ukrainian", "Norwegian", "Japanese");
var Drinks  = List.of("Coffee", "Tea", "Milk", "OrangeJuice", "Water");
var Pets    = List.of("Dog", "Snails", "Horse", "Fox", "Zebra");
var Brands  = List.of("LuckyStrike", "Parliaments", "Kools", "Chesterfields", "OldGold");
var Colours = List.of("Red", "Green", "Ivory", "Yellow", "Blue");
```

```java
CSP zebraCSP() {
    List<String> Constraints = new ArrayList<>();
    Constraints.addAll(allDifferent(Nations));
    Constraints.addAll(allDifferent(Drinks));
    Constraints.addAll(allDifferent(Pets));
    Constraints.addAll(allDifferent(Brands));
    Constraints.addAll(allDifferent(Colours));
    // There are five houses
    var Values = List.of(1, 2, 3, 4, 5);
    // The Englishman lives in the red house.
    // your code here
    // The Spaniard owns the dog.
    // your code here
    // Coffee is drunk in the green house.
    // your code here
    // The Ukrainian drinks tea.
    // your code here
    // The green house is immediately to the right of the ivory house.
    // your code here
    // The Old Gold smoker owns snails.
    // your code here
    // Kools are smoked in the yellow house.
    // your code here
    // Milk is drunk in the middle house.
    // your code here
    // The Norwegian lives in the first house.
    // your code here
    // The man who smokes Chesterfields lives in the house next to the man with the fox.
    // your code here
    // Kools are smoked in the house next to the house where the horse is kept.
    // your code here
    // The Lucky Strike smoker drinks orange juice.
    // your code here
    // The Japanese smokes Parliaments.
    // your code here
    // The Norwegian lives next to the blue house.
    // your code here

    List<String> Variables = List.of(
        "your code here"
    );
    return new CSP(Variables, Values, Constraints);
}
```

## Functions to Print the Solution

```java
void showHTML(Map<String, Integer> solution) {
    var result = new StringBuilder();
    result.append("<table style=\"border:2px solid blue; text-align:center; font-family:sans-serif;\">\n");
    result.append("<tr>");
    for (var name : List.of("House", "Nationality", "Drink", "Animal", "Brand", "Colour")) {
        result.append("<th style=\"color:gold; background-color:blue; padding: 5px 10px;\">" + name + "</th>");
    }
    result.append("</tr>\n");
    for (int house = 1; house <= 5; house++) {
        result.append("<tr><td style=\"border:1px solid green; padding: 5px;\">" + house + "</td>");
        for (var category : List.of(Nations, Drinks, Pets, Brands, Colours)) {
            for (var item : category) {
                if (solution.get(item) == house) {
                    result.append("<td style=\"border:1px solid green; padding: 5px;\">" + item + "</td>");
                }
            }
        }
        result.append("</tr>\n");
    }
    result.append("</table>");
    display(result.toString(), "text/html");
}
```

```java
showHTML(solution);
```

## Checking the Uniqueness

The function `negateSolution` takes a map that represents a solution and returns a formula
that is true iff one of the variables is different than in the solution.

```java
String negateSolution(Map<String, Integer> solution, List<String> vars) {
    return vars.stream().map(v -> v + " != " + solution.get(v)).collect(Collectors.joining(" || "));
}
```

```java
negateSolution(solution, zebra.variables())
```

The function `checkUniqueness` takes two arguments:
* `solution` is a solution of `csp`.
* `csp` is a *constraint satisfaction problem*.

It tries to compute a new solution for the constraint satisfaction problem that is different from the given solution.

```java
Map<String, Integer> checkUniqueness(Map<String, Integer> solution, CSP csp) {
    List<String> newConstraints = new ArrayList<>(csp.constraints());
    newConstraints.add(negateSolution(solution, csp.variables()));
    var newCSP      = new CSP(csp.variables(), csp.values(), newConstraints);
    var newSolution = solve(newCSP);
    if (newSolution != null) {
        System.out.println("The solution is not unique. The alternative solution is:");
        showHTML(newSolution);
        return newSolution;
    } else {
        System.out.println("Well done! The solution is unique.");
        return null;
    }
}
```

```java
long start = System.currentTimeMillis();
checkUniqueness(solution, zebra);
System.out.println("uniquenessCheck: " + (System.currentTimeMillis() - start) + " ms");
```
