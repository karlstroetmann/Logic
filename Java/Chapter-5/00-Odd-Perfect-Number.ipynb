```java
@SafeVarargs
<T> RecursiveSet<T> set(T... elements) {
    return new RecursiveSet<>(elements);
}
```

```java
RecursiveSet<Integer> range(int start, int stop) {
    RecursiveSet<Integer> result = new RecursiveSet<>();
    for (int i = start; i <= stop; ++i) {
        result.add(i);
    }
    return result;
}
```

Given a natural number `n`, the function `divisors(n)` computes the set of all 
natural numbers that divide `n`:
$$ \texttt{divisors}(n) = \{ i \in \mathbb{N} \mid n \,\texttt{%}\, i = 0 \} $$

```java
RecursiveSet<Integer> divisors(int n) {
    return range(1, n - 1).filter(i -> n % i == 0);
}
```

A natural number $n$ is perfect if $n$ is equal to the sum of its proper divisors:
$$ \sum \bigl\{t \mid t \in \{1,\cdots, n-1\} \wedge n \;\%\; t = 0 \bigr\}  = n $$
For example, $6$ is perfect, because $6 = 1 + 2 + 3$, but $8$ is not perfect, because 
$1 + 2 + 4 = 7 \not= 8$.

```java
boolean perfect(int n) {
    return divisors(n).reduce(0, (s, i) -> s + i) == n;
}
```

```java
perfect(6)
```

```java
perfect(8)
```

```java
range(1, 10_000).filter(n -> perfect(n))
```

The function `findOddPerfect` searches for an *odd* perfect number.  Nobody knows whether there is an odd perfect number.  Therefore, the function might run forever.  You can stop it by interrupting the kernel.

```java
int findOddPerfect() {
    int n = 1;
    while (true) {
        if (perfect(n)) {
            return n;
        }
        if (n % 1000 == 1) {
            System.out.println(n);
        }
        n += 2;
    }
}
```
