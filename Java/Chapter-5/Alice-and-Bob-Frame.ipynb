### Solution

In order to solve this problem, we have to formalise using first order logic.  Our first task is to choose 
appropriate variables:
 * `A` is the age of Alice,
 * `B` is the age of Bob,
 * `D` is the age difference between Alice and Bob:
   $$ \texttt{D} = \texttt{A} - \texttt{B} $$

   Obviously, `D` is redundant since it can be computed from `A` and `B`.  Nevertheless, using `D` 
   simplifies the formalisation of the problem.

We load the Java API of `Z3` from *Maven Central*, which is the standard repository for Java libraries.  The magic command `%maven` downloads the library when it is executed for the first time.  Then, we import the package `com.microsoft.z3`.

```java
%maven tools.aqua:z3-turnkey:4.14.1
import com.microsoft.z3.*;
```

All `Z3` objects are created by a *context*.  We call this context `Z3`.

```java
Context Z3 = new Context();
```

We declare our constraint variables for Alice's and Bob's current ages. Since ages in these types of riddles are typically whole numbers, we use `Z3.mkIntConst`.

```java
IntExpr A = Z3.mkIntConst("Alice");
IntExpr B = Z3.mkIntConst("Bob");
```

Next, we create the *solver* object.

```java
Solver S = Z3.mkSolver();
```

```java
// your code here
```

We run the solver to check for satisfiability and extract the model.

```java
var status = S.check();
Model model = S.getModel();
model.toString()
```

Finally, we evaluate the variables, convert them to Java integers, and output the solution.

```java
int ageAlice = Integer.parseInt(model.eval(A, true).toString());
int ageBob   = Integer.parseInt(model.eval(B, true).toString());
```

```java
System.out.println("Alice is " + ageAlice + " years old.");
System.out.println("Bob is " + ageBob + " years old.");
System.out.println("\nVerification:");
System.out.println("The age difference is " + (ageAlice - ageBob) + " years.");
System.out.println("When Alice was " + ageBob + " (" + (ageAlice - ageBob) + " years ago), Bob was " + (ageBob - (ageAlice - ageBob)) + ".");
System.out.println("Is Alice's age (" + ageAlice + ") twice Bob's past age (" + (ageBob - (ageAlice - ageBob)) + ")?");
System.out.println("When Bob is " + ageAlice + " (in " + (ageAlice - ageBob) + " years), Alice will be " + (ageAlice + (ageAlice - ageBob)) + ".");
System.out.println("Will the sum be 90? " + ageAlice + " + " + (ageAlice + (ageAlice - ageBob)) + " = " + (ageAlice * 2 + (ageAlice - ageBob)) + " = 90?");
```
