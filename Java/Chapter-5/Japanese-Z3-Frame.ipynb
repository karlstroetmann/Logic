# Japanese IQ Test

The following puzzle is part of a [recruitment test](https://flashmuseum.org/iq-%E9%81%8E%E6%B2%B3%E9%81%8A%E6%88%B2/) in Japan.

---

A policeman, a convict, a father and his two sons Anton and Bruno, and a mother with her two daughters Cindy and Doris have to cross a river.  On the boat there is only room for two passengers.

During the crossing, the following conditions have to be observed:
* The father is not allowed to be on a shore with one of the daughters 
  if the mother is on the other shore. 
* The mother is not allowed to be on a shore with one of the sons 
  if the father is on the other shore.
* If the criminal is not alone, then the policeman must watch him.

  However the convict can be alone on a shore, as his shackles prevent him from running away. 
* Only the father, the mother, and the policeman are able to steer the boat. 

We are going to formulate this problem as a *symbolic transition system*.
Then, we can solve the problem using our constraint solver.

We will use the following variables to encode the problem:
* `P` equals `1` if the policeman is on the western shore, while `PX` represents the variable `P'`.
* `V` equals `1` if the villain is on the western shore, while `VX` represents the variable `V'`
* `M` equals `1` if the mother is on the western shore, while `MX` represents the variable `M'`
* `F` equals `1` if the father is on the western shore, while `FX` represents the variable `F'`
* `B` is the number of boys on the western shore, while `BX` represents the variable `B'`
* `G` is the number of girls on the western shore, while `GX` represents the variable `G'`.

The function `japaneseCSP(n)` tries to solve the problem with `n` crossings.  If this is possible, it returns a map that maps the variables to their values.  Otherwise, it returns `null`.  The helper function `addAll` adds a list of constraints to the solver.

```java
void addAll(Solver solver, List<BoolExpr> constraints) {
    solver.add(constraints.toArray(BoolExpr[]::new));
}
```

```java
Map<String, Integer> japaneseCSP(int n) {
    Solver solver = Z3.mkSolver();
    List<IntExpr> Ps = range(n).stream().map(i -> Z3.mkIntConst("P" + i)).toList();
    List<IntExpr> Vs = range(n).stream().map(i -> Z3.mkIntConst("V" + i)).toList();
    List<IntExpr> Fs = range(n).stream().map(i -> Z3.mkIntConst("F" + i)).toList();
    List<IntExpr> Ms = range(n).stream().map(i -> Z3.mkIntConst("M" + i)).toList();
    List<IntExpr> Bs = range(n).stream().map(i -> Z3.mkIntConst("B" + i)).toList();
    List<IntExpr> Gs = range(n).stream().map(i -> Z3.mkIntConst("G" + i)).toList();
    addAll(solver, start(Ps.get(0), Vs.get(0), Ms.get(0), Fs.get(0), Bs.get(0), Gs.get(0)));
    addAll(solver, goal (Ps.get(n), Vs.get(n), Ms.get(n), Fs.get(n), Bs.get(n), Gs.get(n)));
    for (int i : range(n - 1)) {
        // your code here
        solver.add(Z3.mkGe(Ps.get(i), num(0)), Z3.mkGe(Vs.get(i), num(0)), Z3.mkGe(Fs.get(i), num(0)),
                   Z3.mkGe(Ms.get(i), num(0)), Z3.mkGe(Bs.get(i), num(0)), Z3.mkGe(Gs.get(i), num(0)),
                   Z3.mkLe(Ps.get(i), num(1)), Z3.mkLe(Vs.get(i), num(1)), Z3.mkLe(Fs.get(i), num(1)),
                   Z3.mkLe(Ms.get(i), num(1)), Z3.mkLe(Bs.get(i), num(2)), Z3.mkLe(Gs.get(i), num(2)));
    }
    if (solver.check() == Status.SATISFIABLE) {
        Model model = solver.getModel();
        Map<String, Integer> result = new LinkedHashMap<>();
        for (var entry : List.of(Map.entry("P", Ps), Map.entry("V", Vs), Map.entry("M", Ms),
                                 Map.entry("F", Fs), Map.entry("B", Bs), Map.entry("G", Gs))) {
            for (int i = 0; i <= n; i++) {
                result.put(entry.getKey() + i, Integer.parseInt(model.eval(entry.getValue().get(i), true).toString()));
            }
        }
        return result;
    }
    return null;
}
```

The function `findSolution` tries to solve the problem with an increasing number of crossings.  It returns a pair consisting of the number `n` of crossings and the solution.

```java
record Result(int n, Map<String, Integer> Solution) {}

Result findSolution() {
    int n = 1;
    while (true) {
        System.out.println("Checking n = " + n + "...");
        var Solution = japaneseCSP(n);
        if (Solution != null) {
            return new Result(n, Solution);
        }
        n += 2;
    }
}
```

```java
long start    = System.currentTimeMillis();
var resultObj = findSolution();
System.out.println("findSolution: " + (System.currentTimeMillis() - start) + " ms");
int n         = resultObj.n();
var Solution  = resultObj.Solution();
System.out.println("\nFound solution in " + n + " steps.");
```

## Auxiliary Code for Pretty Printing

The following code prints the path that has been found to the console, acting as a step-by-step trace showing the boat crossings.

```java
void showSolution(Map<String, Integer> Solution, int n) {
    for (int i = 0; i <= n; i++) {
        int P = Solution.get("P" + i);
        int V = Solution.get("V" + i);
        int M = Solution.get("M" + i);
        int F = Solution.get("F" + i);
        int B = Solution.get("B" + i);
        int G = Solution.get("G" + i);
        String left  = "👮".repeat(P) + "🦹".repeat(V) + "👩".repeat(M) + "👨🏿".repeat(F) + "👦".repeat(B) + "👧🏾".repeat(G);
        String right = "👮".repeat(1 - P) + "🦹".repeat(1 - V) + "👩".repeat(1 - M) + "👨🏿".repeat(1 - F)
                     + "👦".repeat(2 - B) + "👧🏾".repeat(2 - G);
        System.out.println(left + " ".repeat(42) + right);
        if (i < n) {
            int sign = i % 2 == 0 ? 1 : -1;   // the boat goes west to east if i is even
            int PS = sign * (Solution.get("P" + i) - Solution.get("P" + (i + 1)));
            int VS = sign * (Solution.get("V" + i) - Solution.get("V" + (i + 1)));
            int MS = sign * (Solution.get("M" + i) - Solution.get("M" + (i + 1)));
            int FS = sign * (Solution.get("F" + i) - Solution.get("F" + (i + 1)));
            int BS = sign * (Solution.get("B" + i) - Solution.get("B" + (i + 1)));
            int GS = sign * (Solution.get("G" + i) - Solution.get("G" + (i + 1)));
            String boat = "👮".repeat(PS) + "🦹".repeat(VS) + "👩".repeat(MS) + "👨🏿".repeat(FS) + "👦".repeat(BS) + "👧🏾".repeat(GS);
            if (i % 2 == 0) {
                System.out.println(" ".repeat(24) + ">>> 🛶 " + boat + " >>>");
            } else {
                System.out.println(" ".repeat(24) + "<<< 🛶 " + boat + " <<<");
            }
        }
    }
}
```

The function `showAnimatedSolution` displays an animation of the solution.  The animation is implemented in *JavaScript*, which is executed by the browser.  The Java code only inserts the data describing the solution into the HTML code: The placeholders `@uid@`, `@n@`, and `@states@` are replaced by a unique identifier, the number of crossings, and the list of states, respectively.  The list of states is given in *JSON* notation.

```java
void showAnimatedSolution(Map<String, Integer> solution, int n) {
    if (solution == null) {
        System.out.println("No solution provided.");
        return;
    }
    var states = new StringJoiner(", ", "[", "]");
    for (int i = 0; i <= n; i++) {
        states.add("{\"P\": %d, \"V\": %d, \"M\": %d, \"F\": %d, \"B\": %d, \"G\": %d}".formatted(
            solution.get("P" + i), solution.get("V" + i), solution.get("M" + i),
            solution.get("F" + i), solution.get("B" + i), solution.get("G" + i)));
    }
    String uid = Integer.toString(new Random().nextInt(Integer.MAX_VALUE), 36);
    String htmlContent = """
    <div style="font-family: sans-serif; max-width: 800px; margin: 0 auto; text-align: center;">
        <div style="margin-bottom: 15px;">
            <button id="btn-reset-@uid@" style="padding: 8px 16px; cursor: pointer;">Reset</button>
            <button id="btn-prev-@uid@" style="padding: 8px 16px; cursor: pointer;">Previous Step</button>
            <button id="btn-next-@uid@" style="padding: 8px 16px; cursor: pointer; font-weight: bold;">Next Step</button>
            <button id="btn-play-@uid@" style="padding: 8px 16px; cursor: pointer; font-weight: bold; background-color: #a5d6a7; border: 1px solid #2e7d32; border-radius: 4px; margin-left: 10px;">Play</button>
            <span id="label-step-@uid@" style="margin-left: 15px; font-weight: bold; font-size: 16px;">Step 0 of @n@</span>
        </div>
        
        <svg width="800" height="250" style="background-color: #e0f7fa; border: 1px solid #b2ebf2; border-radius: 8px;">
            <rect x="0" y="0" width="250" height="250" fill="#a5d6a7" />
            <rect x="550" y="0" width="250" height="250" fill="#a5d6a7" />
            
            <text x="125" y="30" font-size="14" font-weight="bold" fill="#2e7d32" text-anchor="middle">Western Shore</text>
            <text x="675" y="30" font-size="14" font-weight="bold" fill="#2e7d32" text-anchor="middle">Eastern Shore</text>

            <text id="west-people-@uid@" x="125" y="125" font-size="28" text-anchor="middle"></text>
            <text id="east-people-@uid@" x="675" y="125" font-size="28" text-anchor="middle"></text>

            <g id="boat-@uid@" style="transition: transform 1.5s ease-in-out;">
                <rect x="260" y="105" width="120" height="40" rx="8" fill="#8d6e63" stroke="#5d4037" stroke-width="2"/>
                <text id="boat-p-@uid@" x="320" y="133" font-size="22" text-anchor="middle"></text>
            </g>
        </svg>
    </div>

    <script>
    (function() {
        const states = @states@;
        let currentStep = 0;
        let isAnimating = false;
        let isPlaying = false;
        
        const boat = document.getElementById('boat-@uid@');
        const westPeople = document.getElementById('west-people-@uid@');
        const eastPeople = document.getElementById('east-people-@uid@');
        const boatP = document.getElementById('boat-p-@uid@');
        const stepLabel = document.getElementById('label-step-@uid@');
        const playBtn = document.getElementById('btn-play-@uid@');
        
        // Web Audio API context
        let audioCtx;
        function getAudioCtx() {
            if (!audioCtx) audioCtx = new (window.AudioContext || window.webkitAudioContext)();
            if (audioCtx.state === 'suspended') audioCtx.resume();
            return audioCtx;
        }

        function playMotorSound() {
            const ctx = getAudioCtx();
            const osc = ctx.createOscillator();
            const filter = ctx.createBiquadFilter();
            const gain = ctx.createGain();
            
            osc.type = 'sawtooth';
            osc.frequency.value = 12; 
            
            filter.type = 'lowpass';
            filter.frequency.value = 150;
            gain.gain.setValueAtTime(0, ctx.currentTime);
            gain.gain.linearRampToValueAtTime(0.6, ctx.currentTime + 0.2); 
            
            osc.connect(filter);
            filter.connect(gain);
            gain.connect(ctx.destination);
            osc.start();
            
            return { osc, gain };
        }
        
        function stopMotorSound(sound) {
            if (sound) {
                const ctx = getAudioCtx();
                sound.gain.gain.linearRampToValueAtTime(0, ctx.currentTime + 0.4); 
                setTimeout(() => sound.osc.stop(), 400);
            }
        }

        function playOdeToJoy() {
            const ctx = getAudioCtx();
            
            // --- Simulated Church Reverb Network ---
            const delay1 = ctx.createDelay();
            delay1.delayTime.value = 0.43;
            const delay2 = ctx.createDelay();
            delay2.delayTime.value = 0.27;
            
            const feedback = ctx.createGain();
            feedback.gain.value = 0.25; 
            
            const reverbFilter = ctx.createBiquadFilter();
            reverbFilter.type = 'lowpass';
            reverbFilter.frequency.value = 1000; 
            const reverbMix = ctx.createGain();
            reverbMix.gain.value = 0.4; 

            delay1.connect(reverbFilter);
            reverbFilter.connect(delay2);
            delay2.connect(feedback);
            feedback.connect(delay1);
            delay1.connect(reverbMix);
            delay2.connect(reverbMix);
            reverbMix.connect(ctx.destination);

            const notes = [
                { f: 329.63, d: 0.4 }, { f: 329.63, d: 0.4 }, { f: 349.23, d: 0.4 }, { f: 392.00, d: 0.4 },
                { f: 392.00, d: 0.4 }, { f: 349.23, d: 0.4 }, { f: 329.63, d: 0.4 }, { f: 293.66, d: 0.4 },
                { f: 261.63, d: 0.4 }, { f: 261.63, d: 0.4 }, { f: 293.66, d: 0.4 }, { f: 329.63, d: 0.4 },
                { f: 329.63, d: 0.6 }, { f: 293.66, d: 0.2 }, { f: 293.66, d: 0.8 },
                { f: 329.63, d: 0.4 }, { f: 329.63, d: 0.4 }, { f: 349.23, d: 0.4 }, { f: 392.00, d: 0.4 },
                { f: 392.00, d: 0.4 }, { f: 349.23, d: 0.4 }, { f: 329.63, d: 0.4 }, { f: 293.66, d: 0.4 },
                { f: 261.63, d: 0.4 }, { f: 261.63, d: 0.4 }, { f: 293.66, d: 0.4 }, { f: 329.63, d: 0.4 },
                { f: 293.66, d: 0.6 }, { f: 261.63, d: 0.2 }, 
            ];

            let startTime = ctx.currentTime + 0.1;
            notes.forEach(note => {
                function strikeBell(freqMultiplier, volume, decay) {
                    const osc = ctx.createOscillator();
                    const gain = ctx.createGain();
                    osc.type = 'sine'; 
                    osc.frequency.value = note.f * freqMultiplier;
                    
                    gain.gain.setValueAtTime(0, startTime);
                    gain.gain.linearRampToValueAtTime(volume, startTime + 0.02);
                    gain.gain.exponentialRampToValueAtTime(0.001, startTime + decay);
                    
                    osc.connect(gain);
                    gain.connect(ctx.destination);
                    gain.connect(delay1);
                    osc.start(startTime);
                    osc.stop(startTime + decay + 1);
                }
                strikeBell(1, 0.6, 3.5);
                strikeBell(2.76, 0.15, 1.8);
                startTime += note.d * 0.95; 
            });
        }

        function makeStr(P, V, M, F, B, G) {
            return '👮'.repeat(P) + '🦹'.repeat(V) + '👩'.repeat(M) + '👨🏿'.repeat(F) + '👦'.repeat(B) + '👧🏾'.repeat(G);
        }

        function renderStaticState(step) {
            const s = states[step];
            westPeople.textContent = makeStr(s.P, s.V, s.M, s.F, s.B, s.G);
            eastPeople.textContent = makeStr(1 - s.P, 1 - s.V, 1 - s.M, 1 - s.F, 2 - s.B, 2 - s.G);
            boatP.textContent = '';
            
            boat.style.transform = step % 2 === 0 ? 'translateX(0px)' : 'translateX(170px)';
            stepLabel.textContent = 'Step ' + step + ' of @n@';
        }

        async function animateTransition(fromStep, toStep) {
            if (isAnimating) return;
            isAnimating = true;
            
            const s0 = states[fromStep];
            const s1 = states[toStep];
            
            const pP = Math.abs(s0.P - s1.P);
            const pV = Math.abs(s0.V - s1.V);
            const pM = Math.abs(s0.M - s1.M);
            const pF = Math.abs(s0.F - s1.F);
            const pB = Math.abs(s0.B - s1.B);
            const pG = Math.abs(s0.G - s1.G);
            
            if (fromStep % 2 === 0) { 
                westPeople.textContent = makeStr(s0.P - pP, s0.V - pV, s0.M - pM, s0.F - pF, s0.B - pB, s0.G - pG);
            } else { 
                eastPeople.textContent = makeStr((1 - s0.P) - pP, (1 - s0.V) - pV, (1 - s0.M) - pM, (1 - s0.F) - pF, (2 - s0.B) - pB, (2 - s0.G) - pG);
            }
            
            boatP.textContent = makeStr(pP, pV, pM, pF, pB, pG);
            
            const sound = playMotorSound();
            boat.style.transform = toStep % 2 === 0 ? 'translateX(0px)' : 'translateX(170px)';
            
            await new Promise(r => setTimeout(r, 1500));
            
            stopMotorSound(sound);
            currentStep = toStep;
            renderStaticState(currentStep);
            
            if (currentStep === states.length - 1) {
                playOdeToJoy();
                stopAutoPlay();
            }
            isAnimating = false;
        }

        function stopAutoPlay() {
            isPlaying = false;
            playBtn.textContent = 'Play';
            playBtn.style.backgroundColor = '#a5d6a7';
        }

        async function toggleAutoPlay() {
            if (isPlaying) {
                stopAutoPlay();
                return;
            }
            if (currentStep >= states.length - 1) {
                currentStep = 0;
                renderStaticState(0);
            }
            isPlaying = true;
            playBtn.textContent = 'Pause';
            playBtn.style.backgroundColor = '#ffcc80'; 

            while (isPlaying && currentStep < states.length - 1) {
                await animateTransition(currentStep, currentStep + 1);
                if (isPlaying) await new Promise(r => setTimeout(r, 400));
            }
        }

        function handleManualAction(actionFn) {
            stopAutoPlay();
            actionFn();
        }

        playBtn.addEventListener('click', toggleAutoPlay);
        document.getElementById('btn-next-@uid@').addEventListener('click', () => {
            handleManualAction(() => {
                if (currentStep < states.length - 1) animateTransition(currentStep, currentStep + 1);
            });
        });
        document.getElementById('btn-prev-@uid@').addEventListener('click', () => {
            handleManualAction(() => {
                if (currentStep > 0) animateTransition(currentStep, currentStep - 1);
            });
        });
        document.getElementById('btn-reset-@uid@').addEventListener('click', () => {
            handleManualAction(() => {
                if (!isAnimating) {
                    currentStep = 0;
                    renderStaticState(0);
                }
            });
        });

        renderStaticState(0);
    })();
    </script>
    
    """.replace("@uid@", uid)
       .replace("@n@", String.valueOf(n))
       .replace("@states@", states.toString());
    display(htmlContent, "text/html");
}
```
