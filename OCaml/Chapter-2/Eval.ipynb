# Evaluating Code Given as a String

<samp>Python</samp> provides the function `eval`, which evaluates an expression that is given as a string, and the function `exec`, which executes statements that are given as a string.  For example, after `exec("def f(x): return x + 1")`, the function `f` can be called like any other function.

<samp>OCaml</samp> is a compiled language and has no such functions.  However, the toplevel that runs this notebook is itself written in <samp>OCaml</samp>, and its module `Toploop` can be used in the notebook: The file `setup.ml`, which is loaded in the first cell, adds the *compiler libraries* containing this module to the search path.  The function `Toploop.use_input` reads <samp>OCaml</samp> phrases from a given input and executes them as if they had been typed into the toplevel.  Hence, every definition in this input is available in all cells that are executed afterwards.

The function `my_eval` takes one argument:
- `code` is a string containing <samp>OCaml</samp> code.

It executes the phrases contained in `code`.  The result is `true` if `code` has been executed successfully and `false` if `code` contains an error.  In the latter case, the error message is printed.

**Implementation:** `Toploop.String code` specifies that the phrases are read from the string `code`.  The first argument of `Toploop.use_input` is a *formatter*, i.e. the output channel for the answers of the toplevel and for the error messages.  The formatter `Format.std_formatter` writes to the standard output.

```ocaml
let my_eval (code : string) : bool =
  Toploop.use_input Format.std_formatter (Toploop.String code)
```

The string `t` defines the function `greet_jupyter`, which takes a name and returns a greeting.  It is written as a quoted string `{|...|}`, so that the double quotes inside need not be escaped.

```ocaml
let t : string = {|
let greet_jupyter (name : string) : string =
  "Hello " ^ name ^ ", I am globally available!"
|}
```

When `t` is executed, the toplevel answers with the type of `greet_jupyter`, just as if the definition had been typed into a cell.

```ocaml
my_eval t
```

Now the function `greet_jupyter` can be used like any other function.

```ocaml
greet_jupyter "World"
```
