# <a href="https://en.wikipedia.org/wiki/Legendre%27s_conjecture">Legendre's Conjecture</a>

The function `range` takes two arguments:
- `start` is an integer,
- `stop` is an integer.

It returns the list of all integers from `start` up to and including `stop`.  If `start` is bigger than `stop`, this list is empty.

**Implementation:** The function `List.init` takes the length of the list and a function that computes the element at index `i`.  The function `max` ensures that the length is not negative.

```ocaml
let range (start : int) (stop : int) : int list =
  List.init (max 0 (stop - start + 1)) (fun i -> start + i)
```

```ocaml
range 1 10
```

The function `divisors` takes one argument:
- `k` is a positive natural number.

It returns the list of all natural numbers `t` from the range $\{1, \cdots, k\}$ that divide `k` evenly, i.e. the list of all `t` such that `k mod t = 0`.

```ocaml
let divisors (k : int) : int list =
  range 1 k |> List.filter (fun t -> k mod t = 0)
```

Let us test this function.  The function `List.map string_of_int` converts a list of integers into a list of strings, and the function call `String.concat ", " l` concatenates the strings of the list `l`, separating them by commas.

```ocaml
for i = 1 to 9 do
  Printf.printf "Divisors of %d: %s.\n" i (String.concat ", " (List.map string_of_int (divisors i)))
done
```

The function `is_prime` takes one argument:
- `k` is a positive natural number.

It returns `true` if `k` is a prime number and `false` otherwise.  A number $k$ is prime if and only if the set of its divisors has the form $\{1, k\}$ where $k \not= 1$.  This is the case if and only if the list of its divisors has exactly two elements.

```ocaml
let is_prime (k : int) : bool =
  List.length (divisors k) = 2
```

```ocaml
range 2 100 |> List.filter is_prime
```

More than 200 years ago the French mathematician
<a href="http://de.wikipedia.org/wiki/Adrien-Marie_Legendre">Adrien-Marie Legendre</a>
(1752 - 1833) [conjectured](https://en.wikipedia.org/wiki/Legendre%27s_conjecture)  that there is always a *prime number* between two successive squares bigger than one, i.e. if we denote the set of all *prime numbers* as $\mathbb{P}$, then we have:
$$ \forall n \in \mathbb{N}: \bigl(n \geq 1 \rightarrow \exists p \in \mathbb{P}: n^2 < p < (n+1)^2 \bigr) $$
The question whether this claim is true is still **open**.  For example, there are two prime numbers between $5^2 = 25$ and $6^2 = 36$:

```ocaml
range 25 36 |> List.filter is_prime
```

The function `legendre` takes one argument:
- `n` is a natural number such that $n \geq 1$.

It checks whether there is a prime number $p$ such that
$$ n^2 < p < (n+1)^2 $$
holds.  The function returns `true` if there is a prime number $p$ between $n^2$ and $(n+1)^2$.  Otherwise, `false` is returned.

**Implementation:** The function searches the list of all numbers from $n^2$ up to $(n+1)^2$.  This list also contains the numbers $n^2$ and $(n+1)^2$, but these numbers are not prime.  The function `List.exists` stops as soon as it has found a prime number.

```ocaml
let legendre (n : int) : bool =
  range (n * n) ((n + 1) * (n + 1)) |> List.exists is_prime
```

This is getting tedious, lets automate the testing.

The function `find_counter_example` takes one argument:
- `n` is the natural number where the search for a counter example starts.

The function tries to find a counter example for Legendre's conjecture, i.e. it tries to find a natural number $m \geq n$ such that there is no prime number $p$ between $m^2$ and $(m+1)^2$.  It prints every number that it has checked successfully.  Once a counter example $m$ is found, the function stops and prints the natural number $m$.  Otherwise, the function keeps going until the sun rises in the west.  The function does not return a meaningful result: its return type is `unit`.

**Implementation:** Instead of a `while` loop, the search is implemented as a recursive function: if `legendre n` is `true`, the function calls itself with the argument `n + 1`.  As this recursive call is the last action of the function (a so called *tail call*), the <samp>OCaml</samp> compiler translates the recursion into a loop.  Hence, the recursion does not consume memory on the stack.

```ocaml
let rec find_counter_example (n : int) : unit =
  if legendre n then begin
    Printf.printf "%d\n" n;
    find_counter_example (n + 1)
  end else
    Printf.printf "No prime between %d^2 and %d^2!\n" n (n + 1)
```

```ocaml
find_counter_example 1
```

The function call `find_counter_example 1` will stop if and only if Legendre's conjecture is false.  Therefore, if we had a function that could take our definition of the function `find_counter_example` and a natural number `n` as inputs and that could then decide, whether the function call `find_counter_example n` would eventually terminate, then this function would be able to decide whether Legendre's conjecture is correct.
