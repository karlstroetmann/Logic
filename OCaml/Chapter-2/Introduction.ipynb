# An Introduction to <samp>OCaml</samp>

## Evaluating expressions

<samp>OCaml</samp> is a compiled language.  However, <samp>OCaml</samp> also comes with an interactive *toplevel*, i.e. an interactive shell that reads <samp>OCaml</samp> code piece by piece, evaluates it, and prints the result.  The <samp>OCaml</samp> kernel that runs this notebook is based on this toplevel.  Therefore, expressions can be evaluated directly.  In a *Jupyter*
notebook we just have to type <tt>Ctrl-Enter</tt> in the cell containing the expression.  Instead
of <tt>Ctrl-Enter</tt> we can also use <tt>Shift-Enter</tt>.

````ocaml
1 + 2
````

The toplevel answers with `- : int = 3`.  This answer has three parts: The symbol `-` indicates that the value has no name, `int` is the *type* of the value, and `3` is the value itself.  The toplevel shows the type of every value that it computes.

## Basic Concepts: Type Annotations and Functions

Before introducing the individual data types and language constructs of <samp>OCaml</samp>, this section introduces two basic concepts that will recur throughout the chapter: type annotations and functions.  In addition, the keyword `let` for the definition of variables and *references* for mutable variables are introduced.

**Type Annotations**

<samp>OCaml</samp> is a *statically typed* language: Every variable, every parameter, and the return value of every function has a type that is known at compile time.  A variable is defined with the keyword `let`, which binds a name to a value.  The type of the variable can be given explicitly by a so-called *type annotation*, which is noted with a colon (`:`) after the name of the variable.  Comments are enclosed in `(*` and `*)`.

````ocaml
let name   : string = "Alice"  (* Variable of type string *)
let age    : int    = 25       (* Variable of type int *)
let active : bool   = true     (* Variable of type bool *)
````

For every definition, the toplevel answers with the keyword `val`, followed by the name, the type, and the value of the defined variable.

Usually, type annotations are optional, since <samp>OCaml</samp> supports *type inference*: The compiler derives the type of a variable from the value that is assigned to it.  For example, when we define the variable `email` below, <samp>OCaml</samp> infers that the type of `email` is `string`.

````ocaml
let email = "karl.stroetmann@dhbw.de"
````

In <samp>OCaml</samp>, the symbol `=` has two meanings: In a `let` definition, it binds a name to a value, while everywhere else it is the *equality operator*.  Therefore, the next cell does not reassign the variable to `email`, but compares `email` with `"karlstroetmann@googlemail.com"`.  

````ocaml
email = "karlstroetmann@googlemail.com"
````

**Several Phrases in One Cell**

The definitions and expressions that the toplevel reads one after the other are called *phrases*.  A cell can contain several phrases.  If a definition is followed by an expression, the definition has to be terminated by two semicolons `;;`.  Otherwise, <samp>OCaml</samp> would read the expression as a continuation of the definition.  The cells below contain examples.

**Immutable Variables and References**

Variables defined with `let` are *immutable*: Once a value has been bound to a name, this value cannot be changed.  The operator `:=`, which changes the content of a *reference*, cannot be applied to `pi`, since `pi` is not a reference.  Hence, the next cell yields an error.

````ocaml
let pi = 3.14159265;;   (* immutable *)
pi := 3.0               (* Error: pi is not a reference *)
````

If we need a variable whose value can change, we have to use a *reference*.  The expression `ref v` creates a reference, i.e. a mutable cell whose initial content is the value `v`.  

````ocaml
let counter = ref 0   (* mutable *)
````

The operator `:=` stores a new value in a reference and the prefix operator `!` returns the current content of a reference.  Note that `!` is **not** the negation operator of TypeScript: in <samp>OCaml</samp>, the negation of a truth value is written as `not`.  The toplevel displays a reference as a record with the field `contents`.  Records are discussed later.

In the next cell, the single semicolon `;` separates two expressions that are evaluated one after the other.  The value of the expression `e1; e2` is the value of `e2`.

In functional programming, references are rarely needed.

````ocaml
counter := 1; !counter 
````

**Functions**

A *function* is a named, reusable piece of code that computes a value from its arguments.  In <samp>OCaml</samp>, a function is defined with the keyword `let`, followed by the name of the function and its parameters.  The types of the parameters as well as the return type may be provided with annotations.  The return type is written after the parameter list.  The body of the function follows the symbol `=`.  It is an *expression*, and there is no `return` keyword: the value of the body is the result of the function.  Unlike TypeScript, <samp>OCaml</samp> uses no curly braces to group code, and unlike Python, the indentation of the code has no meaning: it only serves to make the code readable.  If a function does not return a meaningful value, its return type is `unit`.  The type `unit` has only one value, which is written as `()`.

In the following code block, a function named `add` is defined.  It takes two parameters:
- `a` is an integer,
- `b` is an integer.

It returns the sum of `a` and `b`, which is an integer, too.  Note that the parameters are *not* separated by commas: every parameter is enclosed in its own pair of parentheses.

An integer, i.e. an `int` is a 63-bit number.  It is stored as 8 bytes, but the last bit is used by the garbage collector.

````ocaml
let add (a: int) (b: int): int =
  a + b
````

The toplevel answers that `add` has the type `int -> int -> int` and that its value is a function, which is displayed as `<fun>`.  The type of a function is written with arrows: `add` takes an `int`, then another `int`, and returns an `int`.

A function is called by writing its arguments after its name.  The arguments are separated by spaces, too:

````ocaml
add 3 5
````

As the arguments of `add` have to be integers, the next cell yields an error.

````ocaml
add "ab" "cd"
````

A function with two parameters can also be called with just one argument.  The result is a function that still expects the second argument.  This is called *partial application*.  For example, `add 1` is a function that takes an integer and returns its successor.

````ocaml
let inc = add 1
````

````ocaml
inc 41
````

**Anonymous Functions**

A function can also be defined without a name using the keyword `fun`.  To the left of the arrow (`->`) are the parameters, and to the right is the body of the function.  Such an *anonymous function* is also called a *lambda expression*.  It can be given a name by binding it to a variable, as shown below.  The function `sub` takes two integers `a` and `b` and returns their difference `a - b`.

````ocaml
let sub = fun (a: int) (b: int): int -> a - b
````

````ocaml
sub 3 5
````

Anonymous functions are often used for short functions that are passed as arguments to other functions, in particular to the functions `List.filter`, `List.map`, `List.fold_left`, and `List.fold_right`, which are introduced in the section on functional programming.

## Primitive Data Types and Arithmetic

In TypeScript and JavaScript, all numbers are floating-point numbers.  Unlike these languages, <samp>OCaml</samp> distinguishes between integers and floating-point numbers.  The most important primitive data types are:

| **Type** | **Values** | **Literals** |
| :--- | :--- | :--- |
| `int` | integers with 63 bits | `42` |
| `float` | floating-point numbers with 64 bits according to the [IEEE-754 standard](https://en.wikipedia.org/wiki/IEEE_754) | `42.0`, `42.`, `4.2e1` |
| `char` | single characters, more precisely single bytes | `'a'` |
| `string` | sequences of bytes, e.g. text in UTF-8 encoding | `"abc"` |
| `bool` | the truth values `true` and `false` | `true` |
| `unit` | the single value `()` | `()` |

Note that a floating-point literal has to contain a decimal point or an exponent: `42` is an `int`, while `42.0` and `42.` are values of type `float`. The toplevel shows the type of every value it computes.

````ocaml
42
````

````ocaml
42.123
````

The range of the type `int` is limited.  On a 64-bit computer, an `int` has 63 bits, since <samp>OCaml</samp> uses one bit of every machine word internally.  The largest value of type `int` is $2^{62} - 1$ and is called `max_int`.  If this limit is exceeded, the result *wraps around* silently: This is called an *overflow*.

````ocaml
max_int
````

````ocaml
max_int + 1
````

The precision of the type `float` is limited, too.  Integers are only represented exactly as long as they are not bigger than $2^{53}$.  Beyond this limit, rounding errors occur, as shown below.

````ocaml
9007199254740992.0 = 9007199254740993.0
````

To bypass these limitations, we can use the library [Zarith](https://github.com/ocaml/Zarith).  It provides integers of arbitrary size, limited only by the available memory, as well as rational numbers.  A library is loaded with the *directive* `#require`.  Directives start with the character `#`.  They are commands for the toplevel and not part of the language <samp>OCaml</samp>.  The directive `#require "zarith.top"` loads Zarith together with the *printers* that display its numbers in the toplevel.

````ocaml
#require "zarith.top"
````

Zarith consists of *modules*.  A module groups related types and functions.  The module `Z` implements integers of arbitrary size: Their type is `Z.t`, i.e. the type `t` that is defined in the module `Z`.  The functions of the module `Z` are accessed in the same way, e.g. `Z.add` adds two integers of type `Z.t`.  The function `Z.of_int` converts an `int` into a `Z.t`.

````ocaml
Z.of_int 1
````

The module `Z` also defines the usual arithmetic operators for its integers.  An expression of the form `Z.(e)` is a *local open*: inside the parentheses, the names defined in `Z` can be used without the prefix `Z.`.  Therefore, the operators `+`, `-`, `*`, and `/` in `e` refer to the arithmetic operators on `Z.t`.  Furthermore, the prefix operator `~$` is another name of `Z.of_int` and the operator `**` computes a power, where the exponent is an `int`.

````ocaml
let result : Z.t = Z.(~$7 ** 1235)  (* Calculation of extremely large powers *)
````

When two integers are divided, the result is always an integer and hence is *truncated* sometimes.
Note that the result is not **rounded** to the nearest integer, but rather it is *rounded towards zero*.  This is true both for `int` and for `Z.t`.

````ocaml
19 / 4
````

````ocaml
-19 / 4
````

````ocaml
Z.(~$(-19) / ~$4)
````

In arithmetic expressions, <samp>OCaml</samp> never converts numbers of one type into another type automatically.  Hence, we can't mix the types `Z.t` and `int` in arithmetic expressions, and the next cell yields an error.

````ocaml
Z.(~$2 * 3)
````

In cases like these, we have to convert the `int` into a `Z.t` using the function `Z.of_int`, i.e. the operator `~$`.

````ocaml
Z.(~$2 * ~$3)
````

## Output and String Literals

For outputting values to the console, <samp>OCaml</samp> provides the function `print_string`, which prints a string, and the function `print_endline`, which prints a string followed by a newline.  The function `Printf.printf` is more flexible: Its first argument is a *format string*, which can contain *format specifiers* like `%d` for integers, `%f` for floating-point numbers, `%c` for characters, `%s` for strings, and `%b` for truth values.  These format specifiers are replaced by the remaining arguments of `Printf.printf`, and the compiler checks that the types of these arguments match the format specifiers.  Unlike `print_endline`, `Printf.printf` does not append a newline automatically.  Inside a string, a newline is written as `\n`.

The data type for textual data is `string`.  In <samp>OCaml</samp>, string literals are enclosed in double quotes (`"`).  Single quotes (`'`) are used for literals of type `char`, which represent a single character.

The functions `print_endline` and `Printf.printf` return the value `()` of type `unit`.  Therefore, the toplevel answers with `- : unit = ()` after the output.  In general, the toplevel displays the value of every expression at the top level of a cell, so we can see the value of an expression without printing it explicitly.

````ocaml
print_endline "1. Hello, World!"
````

````ocaml
Printf.printf "%d. Hello, World%c\n" 2 '!'
````

````ocaml
(* Implicit output of the value of an expression *)
"Hello, World!"
````

**Formatted Strings**

Unlike TypeScript, <samp>OCaml</samp> has no template strings that embed expressions into a string.  Instead, the function `Printf.sprintf` takes a format string and the values for its format specifiers, just like `Printf.printf`.  However, it returns the resulting string instead of printing it.  Furthermore, strings can be concatenated with the operator `^`, and functions like `string_of_int` and `string_of_float` convert numbers into strings.

Note that the following cell defines the variables `name` and `age` a second time.  These new definitions *shadow* the definitions given above: The old variables are not changed, but they can't be accessed by their names anymore.

````ocaml
let name = "Alice"
let age  = 25
````

````ocaml
(* The values of name and age replace the format specifiers %s and %d *)
Printf.printf "User %s is %d years old.\n" name age
````

````ocaml
(* An argument can be an arbitrary expression *)
Printf.printf "Next year, she will be %d.\n" (age + 1)
````

````ocaml
(* Printf.sprintf returns the formatted string *)
let message = Printf.sprintf "User %s is %d years old." name age
````

````ocaml
(* Concatenation of strings *)
print_endline ("User " ^ name ^ " is " ^ string_of_int age ^ " years old.")
````

**Multi-line Strings**

String literals can span multiple lines.  If a string contains double quotes, it is best written as a *quoted string*, which is enclosed in `{|` and `|}`.  Inside a quoted string, neither `"` nor `\` need to be escaped.

````ocaml
let poem = {|
  I am a teacher by profession,
  mostly really by obsession,
  but even though I boldly try,
  I just cannot teach "pigs" to fly.
|};;
print_endline poem;;
````

## Arithmetic Operations and Variable Assignment

<samp>OCaml</samp> provides the usual arithmetic operators.  A peculiarity of <samp>OCaml</samp> is that integers and floating-point numbers have *different* arithmetic operators: The operators for floating-point numbers end with a dot.

| **Operation** | **`int`** | **`float`** |
| :--- | :--- | :--- |
| addition | `+` | `+.` |
| subtraction | `-` | `-.` |
| multiplication | `*` | `*.` |
| division | `/` | `/.` |
| remainder | `mod` | `Float.rem` |
| exponentiation | | `**` |

The reason for this design is type inference: As every operator has exactly one type, <samp>OCaml</samp> can infer the types of the variables in an arithmetic expression from the operators.

````ocaml
7 / 2        (* integer division *)
````

````ocaml
7.0 /. 2.0   (* floating point division *)
````

````ocaml
7 mod 3      (* remainder *)
````

````ocaml
2.0 ** 10.0  (* exponentiation *)
````

````ocaml
2.0 ** 0.5
````

As the operator `+` can only be applied to integers, the next cell yields an error.

````ocaml
1.0 + 2.0
````

A significant difference compared to  Python concerns division.  In <samp>OCaml</samp>, the operator `/` performs an *integer division*.  To perform a floating-point division of two integers, both have to be converted into floating-point numbers using the function `float_of_int`.  Conversely, the function `int_of_float` converts a floating-point number into an integer by cutting off its fractional part.

Integer division rounds towards zero.  If the result is negative, this differs from rounding down.  The function `floor`, which is discussed in the section on numerical functions, rounds a floating-point number down.  Similarly, the remainder computed by `mod` has the sign of the dividend.

````ocaml
float_of_int 7 /. float_of_int 3   (* Floating point division *)
````

````ocaml
-7 / 2                             (* Integer division, rounds towards zero *)
````

````ocaml
int_of_float (floor (-7.0 /. 2.0))                (* Rounding down *)
````

````ocaml
-7 mod 3                           (* Remainder *)
````

**Recursive Functions**

<samp>OCaml</samp> has no operator for the powers of integers.  However, we can define a function for this purpose.  The function `power` takes two arguments:
- `b` is an integer, the *base*,
- `n` is a natural number, the *exponent*.

It returns the power $b^n$.

**Implementation:** The function is based on the following two equations:
- $b^0 = 1$,
- $b^n = b \cdot b^{n-1}$ if $n > 0$.

Hence, `power` calls itself in order to compute $b^{n-1}$.  A function that calls itself is called *recursive*.  In <samp>OCaml</samp>, a recursive function has to be defined with `let rec` instead of `let`.  The expression `if c then e1 else e2` has the value of `e1` if the condition `c` is `true` and the value of `e2` otherwise.  It is discussed in the section on control structures.

````ocaml
let rec power (b : int) (n : int) : int =
  if n = 0 then 1 else b * power b (n - 1)
````

````ocaml
power 2 10
````

**Local Definitions**

Variables can also be defined *locally* inside an expression.  The expression
```
let x = e1 in e2
```
first evaluates `e1` and binds its value to the variable `x`.  Then it evaluates `e2`, and the value of `e2` is the value of the whole expression.  The variable `x` is only visible inside `e2`.  Local definitions are mostly used inside of functions to name intermediate results.

The function `distance` takes four arguments:
- `x1` and `y1` are the coordinates of a point $p_1 = \langle x_1, y_1 \rangle$ in the plane,
- `x2` and `y2` are the coordinates of a point $p_2 = \langle x_2, y_2 \rangle$.

It returns the Euclidean distance $\sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$ of the points $p_1$ and $p_2$.  The local variables `dx` and `dy` store the differences of the coordinates.  The function `sqrt` computes the square root.

````ocaml
let distance (x1: float) (y1: float) (x2: float) (y2: float): float =
  let dx = x2 -. x1 in
  let dy = y2 -. y1 in
  sqrt (dx *. dx +. dy *. dy)
````

````ocaml
distance 0.0 0.0 3.0 4.0
````

## Lists and Arrays

**Lists**

*Lists* are the most important data structure of functional programming.  A list is a sequence of elements of the same type.  In <samp>OCaml</samp>, a list is written in square brackets "`[`" and "`]`", and its elements are separated by *semicolons* `;`.  The empty list, i.e., a list without elements, is written as "`[]`".  The type of a list of integers is written as `int list`.

````ocaml
[1; 2; 3]
````

Be careful not to separate the elements of a list by commas: In <samp>OCaml</samp>, commas create a *tuple* (see below).  Hence, the following list does not contain three integers, but only a list containing a single element, which is the tuple `(1, 2, 3)`.

````ocaml
[1, 2, 3]
````

**Accessing List Elements**

The function `List.nth` takes a list and an index and returns the element at this index.  <samp>OCaml</samp>, like most modern programming languages, uses zero-based indexing, meaning the first element of a list has the index `0`.  <samp>OCaml</samp> lists are *linked lists*: Every element is stored together with a reference to the rest of the list.  Hence, `List.nth xs n` has to walk through the first `n` elements of `xs`.  Therefore, most functions on lists do not use indices, but process the elements one after the other.

````ocaml
let xs = [1; 2; 3];;
List.nth xs 0  (* Returns the first element *)
````

Lists are *immutable*: Once a list has been created, its elements cannot be changed.  Accessing an index that does not exist raises an *exception*, which aborts the computation.  Unlike TypeScript, <samp>OCaml</samp> has no value `undefined`.  In particular, every variable has to be given a value when it is defined.  Hence, the next cell yields an error.

````ocaml
List.nth xs 10
````

If we are not sure whether an index exists, we can use the function `List.nth_opt`.  It returns a value of the type `int option`: The value `Some x` means that the element at the given index is `x`, while the value `None` means that there is no such element.  The type `option` is the <samp>OCaml</samp> replacement for the values `undefined` and `null`.

````ocaml
List.nth_opt xs 1
````

````ocaml
List.nth_opt xs 10
````

**Concatenating Lists**

Lists can be concatenated using the operator `@`.  This operator creates a new list containing all elements of the first list followed by all elements of the second list.  As lists are immutable, the list `xs` remains unchanged.

````ocaml
[1; 2; 3] @ [4; 5; 6]
````

````ocaml
xs @ [4; 5; 6]
````

````ocaml
xs
````

The operator `::` (pronounced *cons*) creates a new list by putting a single element in front of a list.  This operation is very fast, since the new list shares all its remaining elements with the old list.

````ocaml
0 :: xs
````

In TypeScript, arrays can be combined with the *spread operator* "`...`".  In <samp>OCaml</samp>, the function `List.concat` takes a list of lists and concatenates all of them.

````ocaml
List.concat [[1; 2; 3]; [0]; [4; 5; 6]]
````

The function `List.length` returns the number of elements of a list.

````ocaml
List.length [4; 5; 4]
````

**Arrays**

Besides lists, <samp>OCaml</samp> has *arrays*.  An array is written in the brackets "`[|`" and "`|]`".  The element at index `i` of the array `arr` is accessed as `arr.(i)`.  Unlike lists, arrays are *mutable*: The expression `arr.(i) <- v` stores the value `v` at index `i`.  However, the length of an array is fixed when it is created.

````ocaml
let arr = [|1; 2; 3|];;
arr.(0) <- 7;
arr
````

````ocaml
arr.(3) <- 8
````

The function `Array.length` returns the number of elements of an array.

````ocaml
Array.length arr
````

````ocaml
[1; 2; "Hugo"]
````

## Functional Programming with Lists

The module `List` provides a number of functions that enable *functional programming*.  The most important ones are:
* `List.filter`,
* `List.map`,
* `List.fold_left`,
* `List.exists`, and `List.for_all`.

These functions take a function as their first argument and a list as their last argument.  They are often combined with the *pipe operator* `|>`: The expression `x |> f` is the same as `f x`.  Hence, the list can be written in front of the function that is applied to it.

**The Function `List.filter`**

The function `List.filter` creates a new list that contains only those elements of the original list for which a passed condition is met.  The original list is not modified.  The condition is passed as a function that returns either `true` (element is kept) or `false` (element is discarded) for each element.  Usually, an *anonymous function* is used for such short conditions.

In the following example, only numbers greater than 3 are filtered out of a list of numbers:

````ocaml
let numbers = [1; 2; 3; 4; 5]
let result  = numbers |> List.filter (fun n -> n > 3)
````

The function `fun n -> n > 3` is called internally for each element of the list.  For `n = 1`, `2`, and `3`, it returns `false`, which is why these elements are discarded.  For `n = 4` and `5`, it returns `true`, so they remain in the result list.  The original list `numbers` remains unchanged.

| **Element (n)** | **Condition (n > 3)** | **Result** |
| :--- | :--- | :--- |
| 1 | false | Drop |
| 2 | false | Drop |
| 3 | false | Drop |
| 4 | true | Keep |
| 5 | true | Keep |

**The Function `List.map`**

The function `List.map` transforms each element of a list using a passed function and returns a new list containing the transformed values.  The length of the resulting list is always identical to that of the original list.

In the following example, each element of a list of numbers is doubled:

````ocaml
let numbers = [1; 2; 3; 4]
let result  = numbers |> List.map (fun n -> n * 2)
````

The function `fun n -> n * 2` is called once for each element and returns its doubled value.  The result is a new list where each position contains the transformed value of the corresponding position in the original.  The original list `numbers` remains unchanged.

**The Function `List.fold_left`**

The function `List.fold_left` condenses a list into a single value by applying a function to each element along with a running intermediate result (the so-called *accumulator*).  In TypeScript, this function is called `reduce`.

The call `List.fold_left (fun acc n -> acc + n) 0 numbers` consists of the following parts:
1. The **combining function**: `fun acc n -> acc + n` defines the calculation rule.
2. The **initial value**: The `0` is the initial value of the accumulator for the calculation.
3. The **parameters** of the combining function:
    * `acc` (short for *Accumulator*): Stores the intermediate result of all previous additions.
    * `n`: Is the current element from the list that is currently being processed.

Note that the initial value is the *second* argument of `List.fold_left`, while the list is the last argument.  Using the pipe operator, the call can also be written as `numbers |> List.fold_left (fun acc n -> acc + n) 0`.  The suffix `left` indicates that the elements are processed from left to right.  There is also a function `List.fold_right`, which processes the elements from right to left.

In the following example, the sum of all elements of a list is calculated:

````ocaml
let numbers = [1; 2; 3; 4]
let sum     = numbers |> List.fold_left (fun acc n -> acc + n) 0
````

The accumulator `acc` starts with the value `0`.  In each step, the current element `n` is added to the previous accumulator value, and the result is passed on as the new accumulator value:

| **Step** | **Accumulator (acc)** | **Element (n)** | **New acc** |
| :--- | :--- | :--- | :--- |
| Start | 0 | --- | 0 |
| 1 | 0 | 1 | 1 |
| 2 | 1 | 2 | 3 |
| 3 | 3 | 3 | 6 |
| 4 | 6 | 4 | 10 |

**The Function `List.exists`**

The function `List.exists` checks whether there is at least one element that satisfies a given condition.  For example,
below we check whether `numbers` contains a number divisible by `13`.

````ocaml
let numbers = [15; 30; 60; 39];;
numbers |> List.exists (fun n -> n mod 13 = 0)
````

**The Function `List.for_all`**

The function `List.for_all` checks whether all elements of a list satisfy a given condition.  For example,
below we check whether all numbers contained in `numbers` are even.

````ocaml
let numbers = [2; 4; 6; 8; 10];;
numbers |> List.for_all (fun n -> n mod 2 = 0)
````

**Creating Lists with `List.init`**

Besides listing the elements, lists can be created with the function `List.init`.  This is particularly useful when a list of a specific length needs to be populated with calculated values, for example, to generate a sequence of numbers.

The function `List.init` takes two arguments: first, the desired length of the list, and second, a function that is called for each index and determines the value at that position.  In the following example, a list with the numbers 1 to 10 is created:

````ocaml
List.init 10 (fun i -> i + 1)
````

The index `i` runs from `0` to `9` and is transformed by `i + 1` into the desired values `1` to `10`.

<samp>OCaml</samp> has no special syntax for ranges of numbers.  The function `range` takes two arguments:
- `a` is an integer, the start of the range,
- `b` is an integer, the end of the range.

It returns the list `[a; a+1; ...; b]` of all integers from `a` up to and including `b`.  If `a` is bigger than `b`, this list is empty.  The function `max` returns the bigger of its two arguments.  It is needed since `List.init` raises an exception if the length of the list is negative.

````ocaml
let range (a : int) (b : int) : int list =
  List.init (max 0 (b - a + 1)) (fun i -> a + i)
````

````ocaml
range 1 10
````

<samp>OCaml</samp> has no list comprehensions either.  Instead, lists are transformed with `List.map` and `List.filter`.  For example, the following expression computes the list of the first 10 square numbers:

````ocaml
range 1 10 |> List.map (fun i -> i * i)
````

The following list contains only the even square numbers.

````ocaml
range 1 10 |> List.filter (fun i -> i mod 2 = 0) |> List.map (fun i -> i * i)
````

**Pattern Matching: Head and Tail of a List**

A powerful feature of <samp>OCaml</samp> is *pattern matching*.  It allows unpacking values from a data structure directly into separate variables.  Every non-empty list consists of its first element, the *head*, and the list of the remaining elements, the *tail*.  The pattern `head :: tail` matches every non-empty list: it binds the variable `head` to the first element and the variable `tail` to the list of the remaining elements.  Separating the *head* and the *tail* of a list is a common pattern in functional programming.

The expression `match` compares a value with a sequence of patterns.  Each pattern starts with a vertical bar `|` and is followed by an arrow `->` and the expression that is evaluated if the pattern matches.  The first pattern that matches is used.  Below, the pattern `[]` matches the empty list.  The value of the `match` expression has the type `(int * int list) option`: It is either `Some (head, tail)`, where `(head, tail)` is a pair consisting of an integer and a list of integers, or `None`.

````ocaml
match [10; 20; 30; 40; 50] with
| head :: tail -> Some (head, tail)
| []           -> None
````

The functions `List.hd` and `List.tl` compute the head and the tail of a list directly.  They raise an exception if the list is empty.

````ocaml
List.hd [10; 20; 30; 40; 50]
````

````ocaml
List.tl [10; 20; 30; 40; 50]
````

## Tuples

A *tuple* in <samp>OCaml</samp> groups a fixed number of values, which can have different types.  The elements of a tuple are separated by commas and are usually enclosed in parentheses.  Unlike lists, which are homogeneously typed (e.g., `string list`), tuples allow the storage of a heterogeneous collection of values in a strict order.  This concept is particularly suitable for grouping related data without having to define a type of its own, for example for coordinate pairs `(x, y)` or for functions returning several values.  The type of a tuple is written with the symbol `*` between the types of its elements.

The type `bool` denotes a truth value and can only take the two values `true` or `false`.  A more detailed treatment of boolean values and operators is provided in the section *Boolean Values and Operators*.

In the following example, the name `profile` is defined as an abbreviation of the tuple type `int * string * bool`.  In <samp>OCaml</samp>, the names of types start with a lowercase letter.  The values of this type are tuples with exactly three elements in the order `int`, `string`, and `bool`:

````ocaml
type profile = int * string * bool                 (* Type abbreviation *)
let user_profile : profile = (101, "Alice", true)  (* Order and types must match exactly *)
````

There is no index operator for tuples.  Instead, the elements of a tuple are extracted by *pattern matching*: The left side of a `let` definition can be a pattern.  The *wildcard* `_` matches every value without binding it to a variable.

````ocaml
let (_, user_name, _) = user_profile;;
"Name: " ^ user_name
````

For *pairs*, i.e. tuples with two elements, the functions `fst` and `snd` return the first and the second element.

````ocaml
fst (1, "one")
````

````ocaml
snd (1, "one")
````

## Records

In TypeScript, *objects* group values as *properties*.  In <samp>OCaml</samp>, the counterpart of these objects are *records*.  A record type groups a fixed number of values, the so-called *fields*, each of which has a name and a type.  Unlike the elements of a tuple, the fields of a record are accessed by their names.  The names of the fields start with a lowercase letter.

A record type is defined with the keyword `type`, and its fields are enclosed in curly braces `{}` and separated by semicolons.  The following example shows a record type `student` with basic fields:

````ocaml
type student = {
  first_name           : string;
  matriculation_number : int;
  is_enrolled          : bool;
}
````

A record is created by listing the values of all fields in curly braces.  The compiler infers the type `student` from the names of the fields.

````ocaml
let student = {
  first_name           = "Marvin";
  matriculation_number = 123456;
  is_enrolled          = true;
}
````

**Accessing Fields**

The fields of a record are accessed with the dot notation `record.field`.  The names of the fields are fixed when the record type is defined.  Unlike TypeScript, <samp>OCaml</samp> has no bracket notation `obj[expr]` that computes the name of a field at runtime.

````ocaml
student.first_name
````

The fields of a record are *immutable*: They cannot be changed after the record has been created.  Hence, the next cell yields an error.

````ocaml
student.is_enrolled <- false
````

Instead, a *functional update* `{ r with ... }` creates a new record that differs from the record `r` only in the fields listed after the keyword `with`.  The original record remains unchanged.

````ocaml
let former_student = { student with is_enrolled = false }
````

If a field has to be changeable, it has to be declared with the keyword `mutable`.  The expression `r.field <- v` stores the value `v` in the mutable field `field` of the record `r`.

````ocaml
type course = {
  title                : string;
  mutable participants : int;
}
````

````ocaml
let logic = { title = "Logic"; participants = 60 };;
logic.participants <- 61;
logic.participants
````

**Maps**

If the keys are only known at runtime, we use a *map* instead of a record.  A map stores *key-value pairs*.  In <samp>OCaml</samp>, maps are created by the *functor* `Map.Make`.  A functor is a function that takes a module and returns a new module.  The argument of `Map.Make` is the module `String`, which contains, among other things, a function `compare` that compares strings.  The result is a module of maps whose keys are strings.  These maps are stored as balanced binary search trees.  In this lecture, maps are used, e.g., to represent variable assignments.

The toplevel answers with the *signature* of the new module `StringMap`, i.e. with the list of all types and functions that this module provides.

````ocaml
module StringMap = Map.Make (String)
````

The function `StringMap.of_list` creates a map from a list of pairs.  The toplevel does not show the content of the map, but displays it as `<abstr>`: The type of a map is *abstract*, i.e. its implementation is hidden inside the module `StringMap`.  The function `StringMap.bindings` returns the list of all key-value pairs of a map, ordered by their keys.

````ocaml
let zip_codes = StringMap.of_list [("Konstanz", 78462); ("Mannheim", 68163)]
````

````ocaml
StringMap.bindings zip_codes
````

The function call `StringMap.find k m` returns the value associated with the key `k` in the map `m`.  It raises an exception if `m` contains no such key.

````ocaml
let key = "Mannheim";;
StringMap.find key zip_codes
````

Like lists, maps are immutable: The function call `StringMap.add k v m` returns a new map in which the key `k` is associated with the value `v`.

````ocaml
StringMap.add "Stuttgart" 70173 zip_codes |> StringMap.bindings
````

**Nested Records**

Records can be nested to map complex data structures.  The `university` example demonstrates a hierarchy where the field `location` is itself a record and `departments` is a list of strings.

````ocaml
type location = {
  street   : string;
  city     : string;
  zip_code : int;
}

type university = {
  name        : string;
  location    : location;
  departments : string list;
  is_public   : bool;
}
````

````ocaml
let university = {
  name        = "HTWG Konstanz";
  location    = { street = "Alfred-Wachtel-Str. 8"; city = "Konstanz"; zip_code = 78462 };
  departments = ["Computer Science"; "Architecture"; "Mechanical Engineering"];
  is_public   = true;
}
````

Access to deeper levels is achieved by chaining the dot operator or by combining it with `List.nth`:

````ocaml
(* Accessing nested record *)
university.location.city
````

````ocaml
(* Accessing list within a record *)
List.nth university.departments 1
````

## Variant Types

Often, a value can have one of several different forms.  For example, a geometric shape might be either a circle, which is given by its radius, or a rectangle, which is given by its width and its height.  In <samp>OCaml</samp>, such data are modeled by *variant types*.  The definition of a variant type lists its *constructors*.  Each constructor starts with a vertical bar `|` and has a name starting with a capital letter.  A constructor can carry data, whose type is written after the keyword `of`.

````ocaml
type shape =
  | Circle    of float
  | Rectangle of float * float
````

A value of a variant type is created by applying a constructor to its data:

````ocaml
let shapes = [Circle 1.0; Rectangle (2.0, 3.0)]
````

Functions on variant types are defined by pattern matching.  The function `area` takes one argument:
- `shape` is a shape.

It returns the area of `shape`.  The function has one branch for every constructor of `shape`.  The patterns of these branches bind the data of the constructors to variables.  The compiler warns if a constructor is forgotten.  The constant `Float.pi` is the number $\pi$.  In later chapters, formulas are represented by variant types and most functions on formulas are defined in this way.

````ocaml
let area (shape : shape) : float =
  match shape with
  | Circle r         -> Float.pi *. r *. r
  | Rectangle (w, h) -> w *. h
````

````ocaml
shapes |> List.map area
````

## Boolean Values and Operators

The representation of truth values in <samp>OCaml</samp> is done via the data type `bool`, which can only take the values `true` and `false`.  These values are central for controlling the program flow and for logical decisions.

**Logical Operators and Lazy Evaluation**

<samp>OCaml</samp> supports the classic logical operations conjunction (AND), disjunction (OR), and negation (NOT):
* `a && b` is the conjunction of `a` and `b`,
* `a || b` is the disjunction of `a` and `b`, and
* `not a` is the negation of `a`.  Note that `not` is a function, not an operator.

An essential concept in the evaluation of `&&` and `||` is *lazy evaluation* (often referred to as *short-circuit evaluation* in this context).  This means that the second operand is only evaluated if it is absolutely necessary for the final result:
* For `A && B`, `B` is no longer evaluated if `A` is already `false` (the result is definitely `false` anyway).
* For `A || B`, `B` is no longer evaluated if `A` is already `true` (the result is definitely `true` anyway).

<samp>OCaml</samp> has no operator for the *exclusive or*.  However, for truth values the operator `<>` can be used instead, since `a <> b` is `true` if and only if exactly one of the operands `a` and `b` is `true`.

````ocaml
true && false
````

````ocaml
true || false
````

````ocaml
not true
````

````ocaml
true <> true
````

**Comparison Operators and Equality**

Comparisons are frequently used to generate truth values.  In <samp>OCaml</samp>, the operator `=` checks whether two values are equal and the operator `<>` checks whether they are different.  <samp>OCaml</samp> does not convert strings into numbers or numbers into strings.  Hence, the expression `5 = "5"` is not *well-typed* and is rejected by the compiler.

````ocaml
5 = "5"
````

The operator `=` compares values *structurally*: Two lists are equal if they contain the same elements in the same order, and two records are equal if all their fields are equal.  This holds even if the two values are stored as different objects in memory.

````ocaml
[1; 2; 3] = [1; 2] @ [3]
````

````ocaml
student = { former_student with is_enrolled = true }
````

<samp>OCaml</samp> also has the operators `==` and `!=`.  However, they do **not** compare values: They check whether their operands are the *same object* in memory.  This is called *physical equality*.  For example, the lists `[1; 2; 3]` and `[1; 2] @ [3]` are equal, but they are different objects.  Hence, the next cell yields `false`.  In this lecture, we only use the operators `=` and `<>`.

````ocaml
[1; 2; 3] == [1; 2] @ [3]
````

The classic size comparisons (`<`, `>`, `<=`, `>=`) are available in <samp>OCaml</samp>.  They can be applied to values of every type, e.g. strings are compared lexicographically.  However, both operands must have the same type.

An important difference compared to Python is the chaining of comparisons.  A mathematical expression like $3 > 2 > 1$ is not accepted in <samp>OCaml</samp>, because the result of `3 > 2` is a `bool` and we can't compare a `bool` with an `int`.  Therefore, mathematical range checks must be explicitly linked with the *logical and operator*: `3 > 2 && 2 > 1`.

````ocaml
3 > 2 > 1
````

````ocaml
3 > 2 && 2 > 1
````

## Control Structures

Control structures serve to control the flow of a program.  In <samp>OCaml</samp>, all control structures are *expressions*, i.e. they have a value.

**Branches**

The basic conditional execution is done using `if`, `then`, and `else`.  Unlike TypeScript, the conditions are not placed in parentheses, but they are followed by the keyword `then`.  The conditions must have the type `bool`.  Several conditions are checked one after the other by writing `else if`.

````ocaml
let x = 17;;
if x < 0 then
  print_endline "The number is negative!"
else if x = 0 then
  print_endline "The number is zero."
else
  print_endline "The number is positive."
````

As `if` is an expression, it can be used to compute a value.  In this case, all branches must have values of the same type.  The `else` branch can only be omitted if the `then` branch has the type `unit`.  If a branch consists of several expressions that are separated by `;`, these expressions have to be enclosed in parentheses or in the keywords `begin` and `end`.  Otherwise, only the first expression would belong to the branch.

````ocaml
let sign = if x < 0 then -1 else if x = 0 then 0 else 1
````

**The `match` Expression**

Instead of a `switch` statement, <samp>OCaml</samp> has the `match` expression, which we have already used to split a list into its head and its tail.  It compares a value with several patterns.  Unlike the `switch` statement of TypeScript, `match` does not "fall through": Only the branch of the first matching pattern is evaluated, so no `break` is needed.  A pattern can be a constant, and it can be combined with a condition that is introduced by the keyword `when`.  The pattern `n when n < 0` matches every number `n` that is negative.  The wildcard `_` matches every value.

````ocaml
let y = -5;;
match y with
| 0            -> "The number is zero."
| n when n < 0 -> "The number is negative!"
| _            -> "The number is positive."
````

**Loop Constructs**

<samp>OCaml</samp> provides various loop constructs for iteration.  The value of a loop is `()`.  Therefore, loops are used for their side effects, e.g. to print something or to change a reference or an array.
* The **counting `for` loop** `for i = a to b do ... done` lets the variable `i` run from `a` up to and including `b`.  The loop `for i = b downto a do ... done` counts downwards.

````ocaml
for i = 0 to 2 do
  Printf.printf "%d\n" i
done
````

* <samp>OCaml</samp> has no loop that iterates over the elements of a list.  Instead, the function `List.iter` takes a function `f` and a list `xs` and calls `f` for every element of `xs`.  The function `f` has to return `()`.  Similarly, `Array.iter` calls a function for every element of an array.

````ocaml
let numbers = [1; 2; 3];;
List.iter (fun element -> Printf.printf "%d\n" element) numbers
````

* The **`while` loop** `while c do ... done` executes its body as long as the condition `c` is met.  Here, the loop variable has to be a reference.  The next cell uses local definitions with `let ... in`, so that the variables `arr` and `i` are only defined inside the cell.

````ocaml
let arr = [|1; 2; 3|] in
let i   = ref 0 in
while !i < Array.length arr do
  Printf.printf "%d\n" arr.(!i);
  i := !i + 1
done
````

<samp>OCaml</samp> has no `break` and no `continue`.  Instead of using `continue` to skip the rest of the current iteration, the body of the loop is put into an `if`.

````ocaml
let arr = [|1; 2; 3|] in
for i = 0 to Array.length arr - 1 do
  if i <> 1 then
    Printf.printf "arr.(%d) = %d\n" i arr.(i)
done
````

**Application Example: [Sieve of Eratosthenes](https://en.wikipedia.org/wiki/Sieve_of_Eratosthenes)**

A classic algorithm for determining all prime numbers up to a given limit $n\in\mathbb{N}$ is the Sieve of Eratosthenes.  The implementation in <samp>OCaml</samp> demonstrates the interaction of arrays, the `for` loop as well as the `while` loop, and conditional expressions:

1. Initialization of an array `primes`, where `primes.(i) = i`.
2. Iteration over all numbers $i$ and $j$ to eliminate multiples (i.e., set them to 0).
3. Filtering the remaining numbers $\neq 0$.

An optimized version also uses $\lceil \sqrt{n} \rceil$ as the upper limit for the outer loop and starts the inner loop at $j=i$ to avoid redundant checks.

The function `primes` takes one argument:
- `n` is a natural number.

It returns the list of all prime numbers that are less than or equal to `n`.

**Implementation:**
- The expression `Array.init (n + 1) (fun i -> i)` creates the array `[|0; 1; ...; n|]`.
- As the functions `sqrt` and `ceil` work with floating-point numbers, `n` is converted into a `float` and the result is converted back into an `int`.
- The body of the `if` consists of a local definition of the reference `j` and a `while` loop.  It is enclosed in `begin` and `end`.
- The value of the `for` loop is `()`.  It is followed by `;`, so that the last line of the function computes the result: The function `Array.to_list` converts the array into a list, from which `List.filter` removes the numbers `0` and `1`.

````ocaml
let primes (n : int) : int list =
  (* Initialize array with values 0 to n *)
  let primes = Array.init (n + 1) (fun i -> i) in
  let limit  = int_of_float (ceil (sqrt (float_of_int n))) in
  for i = 2 to limit do
    (* If i is already marked (0), skip it *)
    if primes.(i) <> 0 then begin
      let j = ref i in
      (* Cross out all multiples of i (starting at i*i) *)
      while i * !j <= n do
        primes.(i * !j) <- 0;
        j := !j + 1
      done
    end
  done;
  (* Filter all non-zero values (prime numbers) *)
  primes |> Array.to_list |> List.filter (fun p -> p >= 2)
````

In <samp>OCaml</samp>, the digits of a numeric literal can be grouped with underscores.  The toplevel only displays the first elements of a long list and indicates the rest by `...`.

````ocaml
primes 10_000
````

The directive `#print_length n` tells the toplevel to display up to `n` elements.  Therefore, we see all prime numbers up to 10,000 below.

````ocaml
#print_length 10_000;;
primes 10_000
````
