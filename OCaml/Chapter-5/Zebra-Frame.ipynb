The variables of a CSP have to start with a lower case letter.  Therefore, we write `english`, `oldGold`, `luckyStrike`, etc.  When we display the solution, we will capitalize these names.

```ocaml
let nations = ["english"; "spaniard"; "ukrainian"; "norwegian"; "japanese"]
let drinks  = ["coffee"; "tea"; "milk"; "orangeJuice"; "water"]
let pets    = ["dog"; "snails"; "horse"; "fox"; "zebra"]
let brands  = ["luckyStrike"; "parliaments"; "kools"; "chesterfields"; "oldGold"]
let colours = ["red"; "green"; "ivory"; "yellow"; "blue"]
```

The function `zebra_csp` takes the argument `()`.  It returns a CSP that codes the zebra problem.  When implementing this function it is important to order the variables in a way that variables that are connected to each other by a constraint are tried in succession, for otherwise the CSP solver will take mu$\cdots$uch longer.

```ocaml
let zebra_csp () : csp =
  let constraints =
    all_different nations @ all_different drinks @ all_different pets
    @ all_different brands @ all_different colours
    @ [ (* The Englishman lives in the red house. *)
        failwith "your code here";
        (* The Spaniard owns the dog. *)
        failwith "your code here";
        (* Coffee is drunk in the green house. *)
        failwith "your code here";
        (* The Ukrainian drinks tea. *)
        failwith "your code here";
        (* The green house is immediately to the right of the ivory house. *)
        failwith "your code here";
        (* The Old Gold smoker owns snails. *)
        failwith "your code here";
        (* Kools are smoked in the yellow house. *)
        failwith "your code here";
        (* Milk is drunk in the middle house. *)
        failwith "your code here";
        (* The Norwegian lives in the first house. *)
        failwith "your code here";
        (* The man who smokes Chesterfields lives in the house next to the man with the fox. *)
        failwith "your code here";
        (* Kools are smoked in the house next to the house where the horse is kept. *)
        failwith "your code here";
        (* The Lucky Strike smoker drinks orange juice. *)
        failwith "your code here";
        (* The Japanese smokes Parliaments. *)
        failwith "your code here";
        (* The Norwegian lives next to the blue house. *)
        failwith "your code here" ]
  in
  let variables : string list = failwith "your code here" in
  (* There are five houses *)
  { variables = variables; values = [1; 2; 3; 4; 5]; constraints = constraints }
```

```ocaml
let zebra = zebra_csp ()
```

## Functions to Print the Solution

The function `show_html` takes one argument:
- `solution` is a solution of the zebra puzzle.

It displays the solution as an HTML table.  For every house, the table shows the nationality, drink, animal, brand, and colour of this house.  `String.capitalize_ascii` turns the first letter of a variable into an upper case letter.

```ocaml
let show_html (solution : assignment) : unit =
  let html = Buffer.create 4096 in
  Buffer.add_string html "<table style=\"border:2px solid blue; text-align:center; font-family:sans-serif;\">\n<tr>";
  List.iter
    (fun name -> Buffer.add_string html (Printf.sprintf "<th style=\"color:gold; background-color:blue; padding: 5px 10px;\">%s</th>" name))
    ["House"; "Nationality"; "Drink"; "Animal"; "Brand"; "Colour"];
  Buffer.add_string html "</tr>\n";
  for house = 1 to 5 do
    Buffer.add_string html (Printf.sprintf "<tr><td style=\"border:1px solid green; padding: 5px;\">%d</td>" house);
    List.iter
      (fun category ->
         List.iter
           (fun item ->
              if RMap.find item solution = house then
                Buffer.add_string html
                  (Printf.sprintf "<td style=\"border:1px solid green; padding: 5px;\">%s</td>" (String.capitalize_ascii item)))
           category)
      [nations; drinks; pets; brands; colours];
    Buffer.add_string html "</tr>\n"
  done;
  Buffer.add_string html "</table>";
  ignore (Jupyter_notebook.display "text/html" (Buffer.contents html))
```

```ocaml
match solution with
| Some s -> show_html s
| None   -> print_endline "No solution found."
```

## Checking the Uniqueness

The function `negate_solution` takes two arguments:
- `solution` is a solution of a CSP,
- `vars` is the list of the variables of this CSP.

It returns a formula that is true iff one of the variables has a different value than in the solution.

```ocaml
let negate_solution (solution : assignment) (vars : string list) : string =
  String.concat " || " (List.map (fun v -> Printf.sprintf "%s <> %d" v (RMap.find v solution)) vars)
```

```ocaml
match solution with
| Some s -> print_endline (negate_solution s zebra.variables)
| None   -> ()
```

The function `check_uniqueness` takes two arguments:
* `solution` is a solution of `csp`.
* `csp` is a *constraint satisfaction problem*.

It tries to compute a new solution for the constraint satisfaction problem that is different from the given solution.  If there is such a solution, it is displayed and returned as `Some`; otherwise, `None` is returned.  The expression `{ csp with constraints = ... }` creates a copy of the record `csp` where the field `constraints` has a new value.

```ocaml
let check_uniqueness (solution : assignment) (csp : csp) : assignment option =
  let new_csp = { csp with constraints = csp.constraints @ [negate_solution solution csp.variables] } in
  match solve new_csp with
  | Some new_solution ->
      print_endline "The solution is not unique. The alternative solution is:";
      show_html new_solution;
      Some new_solution
  | None ->
      print_endline "Well done! The solution is unique.";
      None
```

```ocaml
match solution with
| Some s -> ignore (time "uniquenessCheck" (fun () -> check_uniqueness s zebra))
| None   -> ()
```
