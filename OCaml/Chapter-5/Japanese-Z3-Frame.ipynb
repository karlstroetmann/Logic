# Japanese IQ Test

The following puzzle is part of a [recruitment test](https://flashmuseum.org/iq-%E9%81%8E%E6%B2%B3%E9%81%8A%E6%88%B2/) in Japan.

---

A policeman, a convict, a father and his two sons Anton and Bruno, and a mother with her two daughters Cindy and Doris have to cross a river.  On the boat there is only room for two passengers.

During the crossing, the following conditions have to be observed:
* The father is not allowed to be on a shore with one of the daughters 
  if the mother is on the other shore. 
* The mother is not allowed to be on a shore with one of the sons 
  if the father is on the other shore.
* If the criminal is not alone, then the policeman must watch him.

  However the convict can be alone on a shore, as his shackles prevent him from running away. 
* Only the father, the mother, and the policeman are able to steer the boat. 

We are going to formulate this problem as a *symbolic transition system*.
Then, we can solve the problem using our constraint solver.

We will use the following variables to encode the problem:
* `P` equals `1` if the policeman is on the western shore, while `PX` represents the variable `P'`.
* `V` equals `1` if the villain is on the western shore, while `VX` represents the variable `V'`
* `M` equals `1` if the mother is on the western shore, while `MX` represents the variable `M'`
* `F` equals `1` if the father is on the western shore, while `FX` represents the variable `F'`
* `B` is the number of boys on the western shore, while `BX` represents the variable `B'`
* `G` is the number of girls on the western shore, while `GX` represents the variable `G'`.

As the names of OCaml variables have to start with a lower case letter, the OCaml variables are called `p`, `px`, `v`, `vx`, etc.

The function `japanese_csp` takes one argument:
- `n` is the number of crossings.

It creates a solver and adds the constraints for `n` crossings.  If these constraints are satisfiable, it returns `Some solution`, where `solution` is a map from the names of the variables to their values.  Otherwise, `None` is returned.

```ocaml
let japanese_csp (n : int) : (string, int) RMap.t option =
  let solver = new_solver () in
  let vars x = Array.of_list (List.map (fun i -> int_const (Printf.sprintf "%s%d" x i)) (range n)) in
  let ps = vars "P" and vs = vars "V" and fs = vars "F" and ms = vars "M" and bs = vars "B" and gs = vars "G" in
  Z3.Solver.add solver (start ps.(0) vs.(0) ms.(0) fs.(0) bs.(0) gs.(0));
  Z3.Solver.add solver (goal  ps.(n) vs.(n) ms.(n) fs.(n) bs.(n) gs.(n));
  List.iter
    (fun i ->
       Z3.Solver.add solver (failwith "your code here");
       Z3.Solver.add solver
         [ps.(i) >=: (int_val 0); vs.(i) >=: (int_val 0); fs.(i) >=: (int_val 0); ms.(i) >=: (int_val 0); bs.(i) >=: (int_val 0); gs.(i) >=: (int_val 0);
          ps.(i) <=: (int_val 1); vs.(i) <=: (int_val 1); fs.(i) <=: (int_val 1); ms.(i) <=: (int_val 1); bs.(i) <=: (int_val 2); gs.(i) <=: (int_val 2)])
    (range (n - 1));
  match Z3.Solver.check solver [] with
  | Z3.Solver.SATISFIABLE ->
      let model = solver_model solver in
      Some (RMap.of_list
              (List.concat_map
                 (fun (x, xs) -> List.map (fun i -> (Printf.sprintf "%s%d" x i, eval_int model xs.(i))) (range n))
                 [("P", ps); ("V", vs); ("M", ms); ("F", fs); ("B", bs); ("G", gs)]))
  | _ -> None
```

The function `find_solution` takes the argument `()`.  It tries an increasing odd number of crossings until a solution is found.  The result is the pair `(n, solution)`.

```ocaml
let find_solution () : int * (string, int) RMap.t =
  let rec try_crossings n =
    Printf.printf "Checking n = %d...\n" n;
    match japanese_csp n with
    | Some solution -> (n, solution)
    | None          -> try_crossings (n + 2)
  in
  try_crossings 1
```

## Auxiliary Code for Pretty Printing

The following code prints the path that has been found, acting as a step-by-step trace showing the boat crossings.

The function `repeat` takes two arguments:
- `s` is a string,
- `k` is a natural number.

It returns the string that consists of `k` copies of `s`.

The function `show_solution` takes two arguments:
- `solution` is a map from the names of the variables to their values,
- `n` is the number of crossings.

It prints the people on both shores and on the boat for every crossing.

```ocaml
let repeat (s : string) (k : int) : string =
  String.concat "" (List.init k (fun _ -> s))

let show_solution (solution : (string, int) RMap.t) (n : int) : unit =
  let get x i = RMap.find (Printf.sprintf "%s%d" x i) solution in
  let people p v m f b g =
    repeat "👮" p ^ repeat "🦹" v ^ repeat "👩" m ^ repeat "👨🏿" f ^ repeat "👦" b ^ repeat "👧🏾" g
  in
  for i = 0 to n do
    let p = get "P" i and v = get "V" i and m = get "M" i and f = get "F" i and b = get "B" i and g = get "G" i in
    print_endline (people p v m f b g ^ String.make 42 ' ' ^ people (1 - p) (1 - v) (1 - m) (1 - f) (2 - b) (2 - g));
    if i < n then begin
      let d x = get x i - get x (i + 1) in
      if i mod 2 = 0 then
        print_endline (String.make 24 ' ' ^ ">>> 🛶 " ^ people (d "P") (d "V") (d "M") (d "F") (d "B") (d "G") ^ " >>>")
      else
        print_endline (String.make 24 ' ' ^ "<<< 🛶 " ^ people (- d "P") (- d "V") (- d "M") (- d "F") (- d "B") (- d "G") ^ " <<<")
    end
  done
```

The function `show_animated_solution` takes the same arguments as `show_solution`.  It displays an animation of the solution.  As in the notebook `05-Missionaries-STS.ipynb`, the HTML code is stored in the quoted string `template`, which contains the placeholders `__UID__`, `__N__`, and `__STATES__`.  The function `replace_all s pattern by` replaces every occurrence of the string `pattern` in `s` by the string `by`.

```ocaml
#require "jupyter.notebook";;
```

```ocaml
let replace_all (s : string) (pattern : string) (by : string) : string =
  let n = String.length s and k = String.length pattern in
  let buffer = Buffer.create n in
  let rec loop i =
    if i > n - k then
      Buffer.add_string buffer (String.sub s i (n - i))
    else if String.sub s i k = pattern then begin
      Buffer.add_string buffer by;
      loop (i + k)
    end else begin
      Buffer.add_char buffer s.[i];
      loop (i + 1)
    end
  in
  loop 0;
  Buffer.contents buffer

let template = {|
    <div style="font-family: sans-serif; max-width: 800px; margin: 0 auto; text-align: center;">
        <div style="margin-bottom: 15px;">
            <button id="btn-reset-__UID__" style="padding: 8px 16px; cursor: pointer;">Reset</button>
            <button id="btn-prev-__UID__" style="padding: 8px 16px; cursor: pointer;">Previous Step</button>
            <button id="btn-next-__UID__" style="padding: 8px 16px; cursor: pointer; font-weight: bold;">Next Step</button>
            <span id="label-step-__UID__" style="margin-left: 15px; font-weight: bold; font-size: 16px;">Step 0 of __N__</span>
        </div>
        
        <svg width="800" height="250" style="background-color: #e0f7fa; border: 1px solid #b2ebf2; border-radius: 8px;">
            <rect x="0" y="0" width="250" height="250" fill="#a5d6a7" />
            <rect x="550" y="0" width="250" height="250" fill="#a5d6a7" />
            
            <text x="125" y="30" font-size="14" font-weight="bold" fill="#2e7d32" text-anchor="middle">Western Shore</text>
            <text x="675" y="30" font-size="14" font-weight="bold" fill="#2e7d32" text-anchor="middle">Eastern Shore</text>

            <text id="west-people-__UID__" x="125" y="125" font-size="28" text-anchor="middle"></text>
            <text id="east-people-__UID__" x="675" y="125" font-size="28" text-anchor="middle"></text>

            <g id="boat-__UID__" style="transition: transform 1.5s ease-in-out;">
                <rect x="260" y="105" width="120" height="40" rx="8" fill="#8d6e63" stroke="#5d4037" stroke-width="2"/>
                <text id="boat-p-__UID__" x="320" y="133" font-size="22" text-anchor="middle"></text>
            </g>
        </svg>
    </div>

    <script>
    (function() {
        const states = __STATES__;
        let currentStep = 0;
        let isAnimating = false;
        
        const boat = document.getElementById('boat-__UID__');
        const westPeople = document.getElementById('west-people-__UID__');
        const eastPeople = document.getElementById('east-people-__UID__');
        const boatP = document.getElementById('boat-p-__UID__');
        const stepLabel = document.getElementById('label-step-__UID__');
        
        // Web Audio API context for the motor sound
        let audioCtx;
        function playMotorSound() {
            if (!audioCtx) audioCtx = new (window.AudioContext || window.webkitAudioContext)();
            if (audioCtx.state === 'suspended') audioCtx.resume();
            
            const osc = audioCtx.createOscillator();
            const filter = audioCtx.createBiquadFilter();
            const gain = audioCtx.createGain();
            
            osc.type = 'sawtooth';
            osc.frequency.value = 12; 
            
            filter.type = 'lowpass';
            filter.frequency.value = 150;
            gain.gain.setValueAtTime(0, audioCtx.currentTime);
            gain.gain.linearRampToValueAtTime(0.6, audioCtx.currentTime + 0.2); 
            
            osc.connect(filter);
            filter.connect(gain);
            gain.connect(audioCtx.destination);
            osc.start();
            
            return { osc, gain };
        }
        
        function stopMotorSound(sound) {
            if (sound) {
                sound.gain.gain.linearRampToValueAtTime(0, audioCtx.currentTime + 0.4); 
                setTimeout(() => sound.osc.stop(), 400);
            }
        }

        function makeStr(P, V, M, F, B, G) {
            return '👮'.repeat(P) + '🦹'.repeat(V) + '👩'.repeat(M) + '👨🏿'.repeat(F) + '👦'.repeat(B) + '👧🏾'.repeat(G);
        }

        function renderStaticState(step) {
            const s = states[step];
            westPeople.textContent = makeStr(s.P, s.V, s.M, s.F, s.B, s.G);
            eastPeople.textContent = makeStr(1 - s.P, 1 - s.V, 1 - s.M, 1 - s.F, 2 - s.B, 2 - s.G);
            boatP.textContent = '';
            
            // step % 2 === 0 means West (0px). step % 2 !== 0 means East (170px to touch the other shore).
            boat.style.transform = step % 2 === 0 ? 'translateX(0px)' : 'translateX(170px)';
            stepLabel.textContent = 'Step ' + step + ' of __N__';
        }

        async function animateTransition(fromStep, toStep) {
            if (isAnimating) return;
            isAnimating = true;
            
            const s0 = states[fromStep];
            const s1 = states[toStep];
            
            const pP = Math.abs(s0.P - s1.P);
            const pV = Math.abs(s0.V - s1.V);
            const pM = Math.abs(s0.M - s1.M);
            const pF = Math.abs(s0.F - s1.F);
            const pB = Math.abs(s0.B - s1.B);
            const pG = Math.abs(s0.G - s1.G);
            
            if (fromStep % 2 === 0) { 
                // West to East
                westPeople.textContent = makeStr(s0.P - pP, s0.V - pV, s0.M - pM, s0.F - pF, s0.B - pB, s0.G - pG);
            } else { 
                // East to West
                eastPeople.textContent = makeStr((1 - s0.P) - pP, (1 - s0.V) - pV, (1 - s0.M) - pM, (1 - s0.F) - pF, (2 - s0.B) - pB, (2 - s0.G) - pG);
            }
            
            boatP.textContent = makeStr(pP, pV, pM, pF, pB, pG);
            
            const sound = playMotorSound();
            boat.style.transform = toStep % 2 === 0 ? 'translateX(0px)' : 'translateX(170px)';
            
            await new Promise(r => setTimeout(r, 1500));
            
            stopMotorSound(sound);
            currentStep = toStep;
            renderStaticState(currentStep);
            isAnimating = false;
        }

        document.getElementById('btn-next-__UID__').addEventListener('click', () => {
            if (currentStep < states.length - 1) animateTransition(currentStep, currentStep + 1);
        });

        document.getElementById('btn-prev-__UID__').addEventListener('click', () => {
            if (currentStep > 0) animateTransition(currentStep, currentStep - 1);
        });

        document.getElementById('btn-reset-__UID__').addEventListener('click', () => {
            if (!isAnimating) {
                currentStep = 0;
                renderStaticState(0);
            }
        });

        renderStaticState(0);
    })();
    </script>
    |}

let show_animated_solution (solution : (string, int) RMap.t) (n : int) : unit =
  let get x i = RMap.find (Printf.sprintf "%s%d" x i) solution in
  let states =
    "[" ^ String.concat ", "
            (List.map
               (fun i -> Printf.sprintf "{\"P\": %d, \"V\": %d, \"M\": %d, \"F\": %d, \"B\": %d, \"G\": %d}"
                           (get "P" i) (get "V" i) (get "M" i) (get "F" i) (get "B" i) (get "G" i))
               (range n))
    ^ "]"
  in
  let uid  = Printf.sprintf "%07x" (Random.bits ()) in
  let html = replace_all (replace_all (replace_all template "__UID__" uid) "__N__" (string_of_int n))
                         "__STATES__" states in
  ignore (Jupyter_notebook.display "text/html" html)
```
