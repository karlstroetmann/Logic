## Pretty Printing the Solution

The function `show_html` takes one argument:
- `solution` is a solution of the puzzle.

It displays the solution as an HTML table.  For every room, the table shows the brand, car, cancer, and name of its patient.

```ocaml
let show_html (solution : assignment) : unit =
  let html = Buffer.create 4096 in
  Buffer.add_string html "<table style=\"border:2px solid blue\">\n<tr>";
  List.iter
    (fun name -> Buffer.add_string html ("<th style=\"color:gold; background-color:blue\">" ^ name ^ "</th>"))
    ["Room"; "Brand"; "Car"; "Cancer"; "Name"];
  Buffer.add_string html "</tr>\n";
  for room = 1 to 5 do
    Buffer.add_string html (Printf.sprintf "<tr><td style=\"border:1px solid green\">%d</td>" room);
    List.iter
      (fun category ->
         List.iter
           (fun x ->
              if RMap.find x solution = room then
                Buffer.add_string html ("<td style=\"border:1px solid green\">" ^ String.capitalize_ascii x ^ "</td>"))
           category)
      [brands; cars; cancers; names];
    Buffer.add_string html "</tr>\n"
  done;
  Buffer.add_string html "</table>";
  ignore (Jupyter_notebook.display "text/html" (Buffer.contents html))
```

```ocaml
match solution with
| Some s -> show_html s
| None   -> print_endline "No solution found."
```

## Checking the Uniqueness

The function `negate_solution` takes two arguments:
- `solution` is a solution of a CSP,
- `vars` is the list of the variables of this CSP.

It returns a formula that is true iff one of the variables has a different value than in the solution.

```ocaml
let negate_solution (solution : assignment) (vars : string list) : string =
  String.concat " || " (List.map (fun v -> Printf.sprintf "%s <> %d" v (RMap.find v solution)) vars)
```

```ocaml
match solution with
| Some s -> print_endline (negate_solution s csp.variables)
| None   -> ()
```

The function `check_uniqueness` takes two arguments:
* `solution` is a solution of `csp`.
* `csp` is a *constraint satisfaction problem*.

It tries to compute a new solution for the constraint satisfaction problem that is different from the given solution.  If there is such a solution, it is displayed and returned as `Some`; otherwise, `None` is returned.

```ocaml
let check_uniqueness (solution : assignment) (csp : csp) : assignment option =
  let new_csp = { csp with constraints = csp.constraints @ [negate_solution solution csp.variables] } in
  match solve new_csp with
  | Some new_solution ->
      print_endline "The solution is not unique. The alternative solution is:";
      show_html new_solution;
      Some new_solution
  | None ->
      print_endline "Well done! The solution is unique.";
      None
```

```ocaml
match solution with
| Some s -> ignore (time "uniquenessCheck" (fun () -> check_uniqueness s csp))
| None   -> ()
```
