The module `RSet`, which is loaded by `setup.ml`, implements sets.  The function `range` takes two arguments:
- `start` is an integer,
- `stop` is an integer.

It returns the set of all integers from `start` up to and including `stop`.

```ocaml
let range (start : int) (stop : int) : int RSet.t =
  RSet.of_list (List.init (max 0 (stop - start + 1)) (fun i -> start + i))
```

The function `divisors` takes one argument:
- `n` is a natural number.

It computes the set of all natural numbers less than `n` that divide `n`:
$$ \texttt{divisors}(n) = \{ i \in \{1, \cdots, n-1\} \mid n \;\texttt{mod}\; i = 0 \} $$

```ocaml
let divisors (n : int) : int RSet.t =
  RSet.filter (fun i -> n mod i = 0) (range 1 (n - 1))
```

```ocaml
divisors 24
```

A natural number $n$ is *perfect* if $n$ is equal to the sum of its proper divisors:
$$ \sum \bigl\{t \mid t \in \{1,\cdots, n-1\} \wedge n \;\texttt{mod}\; t = 0 \bigr\}  = n $$
For example, $6$ is perfect, because $6 = 1 + 2 + 3$, but $8$ is not perfect, because
$1 + 2 + 4 = 7 \not= 8$.

The function `perfect` takes one argument:
- `n` is a natural number.

It returns `true` if `n` is perfect.  `RSet.fold ( + ) s 0` computes the sum of the elements of the set `s`.  Here, the operator `+` is written as the function `( + )`.

```ocaml
let perfect (n : int) : bool =
  RSet.fold ( + ) (divisors n) 0 = n
```

```ocaml
perfect 6
```

```ocaml
perfect 8
```

```ocaml
RSet.filter perfect (range 1 10_000)
```

Nobody knows whether there is an odd perfect number.  The function `find_odd_perfect` takes the argument `()` and searches for an odd perfect number.  It returns the first odd perfect number it finds.  Every 1000 numbers, it prints the number it is testing, so that we can see its progress.

As nobody knows whether this function terminates, the cell calling it is tagged `skip-execution`.  If you run it, you have to interrupt the kernel.

```ocaml
let find_odd_perfect () : int =
  let n = ref 1 in
  while not (perfect !n) do
    if !n mod 1000 = 1 then Printf.printf "%d\n" !n;
    n := !n + 2
  done;
  !n
```

```ocaml
find_odd_perfect ()
```
