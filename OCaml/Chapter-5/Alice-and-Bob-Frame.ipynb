# The Alice and Bob Age Problem

In this notebook we are going to solve the following text problem using the constraint solver `Z3`.
<ul>
<li>    Alice is twice as old as Bob was when Alice was as old as Bob is now.</li>
<li>    When Bob is as old as Alice is now, the sum of their ages will be 90.</li>
    
    How old are Alice and Bob now?

### Solution

In order to solve this problem, we have to formalise using first order logic.  Our first task is to choose 
appropriate variables:
 * `A` is the age of Alice,
 * `B` is the age of Bob,
 * `D` is the age difference between Alice and Bob:
   $$ \texttt{D} = \texttt{A} - \texttt{B} $$

   Obviously, `D` is redundant since it can be computed from `A` and `B`.  Nevertheless, using `D` 
   simplifies the formalisation of the problem.

We declare our constraint variables for Alice's and Bob's current ages.  Since ages in these types of riddles are typically whole numbers, we use the function `int_const`.  As names of <samp>OCaml</samp> variables have to start with a lower case letter, we use `a` and `b`.  In the constraints, which are written as strings, the variables are referred to by the names `Alice` and `Bob` that are given to `int_const`.

```ocaml
let a = int_const "Alice"
let b = int_const "Bob"
```

```ocaml
let model = solver_model s
```

Finally, we evaluate the variables, convert them to <samp>OCaml</samp> integers, and output the solution.

```ocaml
let age_alice = eval_int model a
let age_bob   = eval_int model b
```
