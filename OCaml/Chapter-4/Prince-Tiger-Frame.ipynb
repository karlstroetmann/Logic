# The Prince and the Tiger

Once upon a time, there was a king who wanted to marry his daughter to a prince. He had it proclaimed throughout the land that he was seeking a husband for his daughter. One day, a prince came by to apply for the position.

Since the king did not want to marry his daughter to some dullard, he led the prince into a room with nine doors. The king told the prince that the princess was in one of the rooms, but that there were other rooms behind which hungry tigers were waiting. Some rooms were also empty. If the prince opened a door with a tiger behind it, it would probably be his last mistake.

The king went on to say that there were signs on all the doors with statements on them. These statements behave as follows:

* In the rooms where there is a tiger, the statement on the sign is false.
* In the room where the princess is, the statement is true.
* With regards to the empty rooms, the situation is a bit more complicated, because there are two possibilities:
    - Either **all** the inscriptions on the empty rooms are true,
    - or **all** the inscriptions on the empty rooms are false.

The prince then read the inscriptions. These were as follows:

1. Room: The princess is in a room with an odd number. There is no tiger in the rooms with an even number.
2. Room: This room is empty.
3. Room: The inscription on Room No. 5 is true, the inscription on Room No. 7 is false, and there is a tiger in Room No. 3.
4. Room: The inscription on Room No. 1 is false, there is no tiger in Room No. 8, and the inscription on Room No. 9 is true.
5. Room: If the inscription on Room No. 2 or on Room No. 4 is true, then there is no tiger in Room No. 1.
6. Room: The inscription on Room No. 3 is false, the princess is in Room No. 2, and there is no tiger in Room No. 2.
7. Room: The princess is in Room No. 1 and the inscription on Room No. 5 is true.
8. Room: There is no tiger in this room and Room No. 9 is empty.
9. Room: Neither in this room nor in Room No. 1 is there a tiger, and moreover, the inscription on Room No. 6 is true.

## Setting up the Required Notebooks

The module `RSet`, which is loaded by `setup.ml`, implements sets that can be nested.  The function `range` takes two arguments:
- `start` is an integer,
- `stop` is an integer.

It returns the set of all integers from `start` up to and including `stop`.

```ocaml
let range (start : int) (stop : int) : int RSet.t =
  RSet.of_list (List.init (stop - start + 1) (fun i -> start + i))
```

Next, we import the Davis-Putnam solver.  The notebook `06-Davis-Putnam.ipynb` imports the notebook `04-CNF.ipynb`, which provides the function `normalize` and imports our propositional logic parser.

```ocaml
import_notebook "06-Davis-Putnam.ipynb"
```

The function `parse_knf` takes one argument:
- `s` is a string representing a formula.

It returns a set of clauses that is equivalent to the formula represented by `s`.

```ocaml
let parse_knf (s : string) : cnf =
  normalize (parse s)
```

## Auxiliary Functions

The functions defined below make it convenient to create the propositional variables `Pi`, `Ti`, and `Zi` for $i \in \{1,\cdots,9\}$:
* `Pi` is true if there is a princess in room `i`.
* `Ti` is true if there is a tiger in room `i`.
* `Zi` is true if the label at the door of room `i` is true.

Besides these variables we will also use the propositional variable `e`.  This variable will be true if the labels on all empty rooms are true.

The functions `p`, `t`, and `z` take one argument:
- `i` is the number of a room.

They return the names of the variables `Pi`, `Ti`, and `Zi`, respectively.

```ocaml
let p (i : int) : string = Printf.sprintf "P%d" i
let t (i : int) : string = Printf.sprintf "T%d" i
let z (i : int) : string = Printf.sprintf "Z%d" i
```

The function `at_most_one` takes one argument:
- `s` is a set of propositional variables.

It returns a set of clauses expressing the fact that at most one of the variables of `s` is true.

```ocaml
let at_most_one (s : string RSet.t) : cnf =
  RSet.flat_map
    (fun p -> RSet.filter_map (fun q -> if p <> q then Some (RSet.of_list [Neg p; Neg q]) else None) s)
    s
```

## Generating the Set of Clauses Describing the Problem

The function `compute_clauses` takes the argument `()`.  It returns the set of clauses that describes the problem.

As it can be used in place of an expression of any type, it serves as a placeholder.  The function `parse_knf` is useful to compute the clauses.  The local function `room_clauses` takes the number `i` of a room and returns the clauses describing the relation between the variables `Pi`, `Ti`, `Zi`, and `e`.

```ocaml
let compute_clauses () : cnf =
  let room_clauses (i : int) : cnf =
    RSet.unions [
      (* In the room containing the princess, the label at the door is true. *)
      failwith "your code here";
      (* In those rooms where there are tigers, the label is false. *)
      failwith "your code here";
      (* The variable e is true if the labels of ALL empty rooms are true.
         Either all labels of empty rooms are true or all those labels are false. *)
      failwith "your code here"
    ]
  in
  let clues = [
    (* 1. Room: The princess is in a room with an odd number.
                There is no tiger in the rooms with an even number. *)
    failwith "your code here";
    (* 2. Room: This room is empty. *)
    failwith "your code here";
    (* 3. Room: The inscription on Room No. 5 is true,
                the inscription on Room No. 7 is false,
                and there is a tiger in Room No. 3. *)
    failwith "your code here";
    (* 4. Room: The inscription on Room No. 1 is false,
                there is no tiger in Room No. 8,
                and the inscription on Room No. 9 is true. *)
    failwith "your code here";
    (* 5. Room: If the inscription on Room No. 2 or 4 is true,
                then there is no tiger in Room No. 1. *)
    failwith "your code here";
    (* 6. Room: The inscription on Room No. 3 is false,
                the princess is in Room No. 2,
                and there is no tiger in Room No. 2. *)
    failwith "your code here";
    (* 7. Room: The princess is in Room No. 1 and
                the inscription on Room No. 5 is true. *)
    failwith "your code here";
    (* 8. Room: There is no tiger in this room and Room No. 9 is empty. *)
    failwith "your code here";
    (* 9. Room: Neither in this room nor in Room No. 1 is there a tiger,
                and the inscription on Room No. 6 is true. *)
    failwith "your code here"
  ] in
  RSet.unions ([
    (* There is a room containing the princess. *)
    failwith "your code here";
    (* There is at most one princess. *)
    failwith "your code here";
    RSet.flat_map room_clauses (range 1 9)
  ] @ List.map parse_knf clues)
```

## Evaluating and Checking the Solution

We extract the solution from the final simplified clauses by looking for the unit clause $\{\texttt{P}i\}$ that survived.

The function `get_solution` takes one argument:
- `solution` is a set of unit clauses.

It returns `Some v` if `solution` contains a unit clause $\{v\}$ where `v` is a variable that starts with the letter `P`.  The function `arb` from the notebook `06-Davis-Putnam.ipynb` returns the only element of a unit clause.  Otherwise, it returns `None`.  `List.find_map f l` returns the first result of the form `Some x` of applying `f` to the elements of `l`, or `None` if there is no such result.

```ocaml
let get_solution (solution : cnf) : string option =
  List.find_map
    (fun unit_clause ->
       match arb unit_clause with
       | Pos v when v.[0] = 'P' -> Some v
       | _                      -> None)
    (RSet.to_list solution)
```

```ocaml
let princess = get_solution solution;;
match princess with
| Some v -> Printf.printf "Die Prinzessin ist im Zimmer Nummer %s.\n" (String.sub v 1 (String.length v - 1))
| None   -> print_endline "There is no princess."
```

We must verify uniqueness.  If the solution is not unique, there is ambiguity in how our problem was constrained.  The solution should be unique.

The function `check_uniqueness` takes two arguments:
- `clauses` is the set of clauses describing the problem,
- `princess` is the variable stating where the princess is.

It adds the unit clause stating that the princess is not in this room to `clauses`.  If the resulting set of clauses is unsatisfiable, the solution is unique.

```ocaml
let check_uniqueness (clauses : cnf) (princess : string) : unit =
  let extended_clauses = RSet.add (RSet.singleton (Neg princess)) clauses in
  let alternative = solve extended_clauses in
  if alternative = RSet.singleton RSet.empty then
    print_endline "Well done: The solution is unique."
  else
    print_endline "ERROR: The solution is not unique."
```

```ocaml
match princess with
| Some v -> check_uniqueness clauses v
| None   -> print_endline "There is no solution."
```
