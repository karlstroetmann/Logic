# A Functional Implementation of `power`

As the powers of numbers get big fast, we use the module `Z` of the library Zarith, which implements integers of arbitrary size.  Their type is `Z.t`.

```ocaml
#require "zarith.top";;
```

The function `range` takes one argument:
- `n` is a natural number.

It returns the list `[1; 2; ...; n]`.

```ocaml
let range (n : int) : int list =
  List.init n (fun i -> i + 1)
```

The function `power` takes two arguments:
- `m` is an integer of type `Z.t`, the *base*,
- `n` is a natural number, the *exponent*.

It returns the power $m^n$.

**Implementation:** The function multiplies the number $1$ with `m` once for every element of the list `range n`, i.e. it computes
$$ m^n = \underbrace{m \cdot m \cdot \cdots \cdot m}_{n}. $$
The elements of the list are not used.  Therefore, the second parameter of the combining function of `List.fold_left` is the wildcard `_`.  `Z.one` is the number $1$ of type `Z.t`, and inside `Z.(...)` the operator `*` denotes the multiplication of numbers of type `Z.t`.

```ocaml
let power (m : Z.t) (n : int) : Z.t =
  List.fold_left (fun acc _ -> Z.(acc * m)) Z.one (range n)
```

```ocaml
power (Z.of_int 3) 4
```
