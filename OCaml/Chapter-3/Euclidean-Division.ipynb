# Euclidean Division

The function `div_mod` takes two arguments:
- `m` is a natural number,
- `n` is a natural number such that $n > 0$.

It returns the unique pair $(q, r)$
that satisfies the following two formulas:
1. $m = q \cdot n + r$,
2. $r < n$.

Hence $q$ is the *quotient* and $r$ is the *remainder* of the division of $m$ by $n$.  The function uses only division by $2$, which is a cheap operation for binary numbers.

**Implementation:** If $m < n$, then $q = 0$ and $r = m$.  Otherwise, the quotient $q'$ and the remainder $r'$ of the division of $m \;\texttt{/}\; 2$ by $n$ are computed recursively.  Here, $m \;\texttt{/}\; 2$ denotes the integer division of $m$ by $2$.  We have $m = 2 \cdot (m \;\texttt{/}\; 2) + m \;\texttt{mod}\; 2$ and therefore
$$ m = 2 \cdot (q' \cdot n + r') + m \;\texttt{mod}\; 2 = 2 \cdot q' \cdot n + 2 \cdot r' + m \;\texttt{mod}\; 2. $$
- If $2 \cdot r' + m \;\texttt{mod}\; 2 < n$, then $q = 2 \cdot q'$ and $r = 2 \cdot r' + m \;\texttt{mod}\; 2$.
- Otherwise, we have $n \leq 2 \cdot r' + m \;\texttt{mod}\; 2 < 2 \cdot n$ and therefore $q = 2 \cdot q' + 1$ and $r = 2 \cdot r' + m \;\texttt{mod}\; 2 - n$.

The result is a pair of integers, which has the type `int * int`.  The pattern `(q, r)` in the local definition `let (q, r) = ... in` extracts the components of the pair computed by the recursive call.  We test the function with numbers up to $10^{12}$.  As an `int` has 63 bits, these numbers fit into the type `int`.

```ocaml
let rec div_mod (m : int) (n : int) : int * int =
  if m < n then
    (0, m)
  else
    let (q, r) = div_mod (m / 2) n in
    if 2 * r + m mod 2 < n then
      (2 * q,     2 * r + m mod 2)
    else
      (2 * q + 1, 2 * r + m mod 2 - n)
```

We test the function `div_mod` using random numbers.  The function call `Random.init 42` initializes the random number generator with the *seed* `42`.  Using a fixed seed ensures that the tests are reproducible.  The function call `Random.full_int b` returns a random number $x$ such that $0 \leq x < b$.

The function `test_div_mod` takes no arguments, i.e. its only argument is `()`.  It computes `div_mod m n` for one million pairs of random numbers $m$ and $n$ such that $0 \leq m < 10^{12}$ and $1 \leq n < 10^{9}$ and checks that the result satisfies the specification of `div_mod`.  If a test fails, it raises an exception: The function `failwith` raises the exception `Failure` with the given message.  If all tests are successful, a message is printed.

```ocaml
let test_div_mod () : unit =
  Random.init 42;
  for _ = 1 to 1_000_000 do
    let m      = Random.full_int 1_000_000_000_000 in
    let n      = 1 + Random.full_int (1_000_000_000 - 1) in
    let (q, r) = div_mod m n in
    if m <> q * n + r || r >= n then
      failwith (Printf.sprintf "Failed for m=%d, n=%d, q=%d, r=%d" m n q r)
  done;
  print_endline "All tests have been successful!"
```

```ocaml
test_div_mod ()
```
