# Recursive Computation of Squares

The function `square` takes one argument:
- `n` is a natural number.

It returns the square $n^2$ without using multiplication.

**Implementation:** The function is based on the following two equations:
1. $0^2 = 0$,
2. $n^2 = (n-1)^2 + 2 \cdot n - 1$ if $n > 0$.

The second equation holds since $(n-1)^2 = n^2 - 2 \cdot n + 1$.  The expression `2 * n` is just an abbreviation of `n + n`.

```ocaml
let rec square (n : int) : int =
  if n = 0 then
    0
  else
    square (n - 1) + 2 * n - 1
```

```ocaml
for i = 0 to 11 do
  Printf.printf "square(%d) = %d\n" i (square i)
done
```
