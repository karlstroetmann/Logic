# Recursive Computation of the Binomial Coefficients

The *binomial coefficient* ${n \choose k}$ is defined as follows:
$$ {n \choose k} = \frac{n!}{k! \cdot (n-k)!} $$
The function `binomial` takes two arguments:
- `n` is a natural number,
- `k` is a natural number such that $k \leq n$.

The function returns the binomial coefficient ${n \choose k}$.

**Implementation:** The function is based on the following recursive equations:
1. ${n \choose 0} = 1$ and ${n \choose n} = 1$,
2. ${n \choose k} = {n-1 \choose k-1} + {n-1 \choose k}$ if $0 < k < n$.

```ocaml
let rec binomial (n : int) (k : int) : int =
  if k = 0 || k = n then
    1
  else
    binomial (n - 1) (k - 1) + binomial (n - 1) k
```

In order to test the function `binomial`, we compare it with the function `Z.bin` from the library Zarith: `Z.bin n k` computes ${n \choose k}$ for a number `n` of type `Z.t` and a number `k` of type `int`.  The function `Z.to_int` converts the result into an `int`.

The expression `assert c` raises the exception `Assert_failure` if the condition `c` is `false`.  Hence the next cell runs without output if all tests are successful.

```ocaml
#require "zarith.top";;
```

```ocaml
for n = 0 to 11 do
  for k = 0 to n do
    assert (binomial n k = Z.to_int (Z.bin (Z.of_int n) k))
  done
done
```

Finally, we print *Pascal's triangle*.  The function `print_newline` ends the current line.

```ocaml
for n = 1 to 9 do
  for k = 0 to n do
    Printf.printf "%d " (binomial n k)
  done;
  print_newline ()
done
```
