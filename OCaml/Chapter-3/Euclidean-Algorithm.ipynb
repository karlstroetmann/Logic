# The Euclidean Algorithm for Computing the Greatest Common Divisor

The function `gcd` takes two arguments:
- `a` is a natural number,
- `b` is a natural number.

The function returns the *greatest common divisor* of `a` and `b`.

**Implementation:** The function is based on the equations $\texttt{gcd}(a, 0) = a$ and $\texttt{gcd}(a, b) = \texttt{gcd}(b, a \;\texttt{mod}\; b)$ for $b \not= 0$.  As long as $b \not= 0$, the pair $(a, b)$ is replaced by the pair $(b, a \;\texttt{mod}\; b)$.  This does not change the greatest common divisor.

In OCaml, the parameters of a function can not be changed.  Therefore, we introduce the *references* `x` and `y` that are initialized with `a` and `b`.  OCaml has no simultaneous assignment like `[a, b] = [b, a % b]` in TypeScript.  Therefore, we store the remainder in the auxiliary variable `r` before `x` is changed.

```ocaml
let gcd (a : int) (b : int) : int =
  let x = ref a in
  let y = ref b in
  while !y <> 0 do
    let r = !x mod !y in
    x := !y;
    y := r
  done;
  !x
```

```ocaml
gcd 120 20
```
