# The Euclidean Algorithm for Computing the Greatest Common Divisor

The function `gcd` takes two arguments:
- `a` is a natural number,
- `b` is a natural number.

The function returns the *greatest common divisor* of `a` and `b`.

**Implementation:** The function is based on the following two equations:
- $\texttt{gcd}(a, 0) = a$,
- $\texttt{gcd}(a, b) = \texttt{gcd}(b, a \;\texttt{mod}\; b)$ if $b \not= 0$.

```ocaml
let rec gcd (a : int) (b : int) : int =
  if b = 0 then
    a
  else
    gcd b (a mod b)
```

```ocaml
gcd 120 20
```
