# Iterative Computation of Squares

The function `square` takes one argument:
- `n` is a natural number.

It returns the square $n^2$ without using multiplication.

**Implementation:** The function adds the first $n$ odd numbers, since
$$ \sum\limits_{i=1}^{n} (2 \cdot i - 1) = n^2. $$
In OCaml, the parameters of a function can not be changed.  Therefore, the function uses the *references* `s`, `r`, and `u`:
- `s` counts down from `n` to `0`,
- `r` is the sum of the odd numbers added so far,
- `u` is the next odd number that is to be added.

The loop maintains the invariants $r = (n - s)^2$ and $u = 2 \cdot (n - s) + 1$.  When the loop terminates, we have $s = 0$ and hence $r = n^2$.

```ocaml
let square (n : int) : int =
  let s = ref n in
  let r = ref 0 in
  let u = ref 1 in
  while !s > 0 do
    r := !r + !u;
    u := !u + 2;
    s := !s - 1
  done;
  !r
```

```ocaml
for i = 0 to 11 do
  Printf.printf "square(%d) = %d\n" i (square i)
done
```
